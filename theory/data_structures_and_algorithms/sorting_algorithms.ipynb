## 1. Sorting Concepts (Stability, In-Place, Complexity Matrix)

- **Stability**: A sorting algorithm is **stable** if it preserves the relative order of elements with equal keys.
- **In-Place**: Requires $O(1)$ or $O(\log n)$ extra memory space.

### Complexity Matrix

| Algorithm | Best Case | Average Case | Worst Case | Space | Stable? | In-Place? |
|-----------|-----------|--------------|------------|-------|---------|-----------|
| **Bubble Sort** | $O(n)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ | Yes | Yes |
| **Selection Sort** | $O(n^2)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ | No | Yes |
| **Insertion Sort** | $O(n)$ | $O(n^2)$ | $O(n^2)$ | $O(1)$ | Yes | Yes |
| **Merge Sort** | $O(n \log n)$ | $O(n \log n)$ | $O(n \log n)$ | $O(n)$ | Yes | No |
| **Quick Sort** | $O(n \log n)$ | $O(n \log n)$ | $O(n^2)$ | $O(\log n)$ | No | Yes |
| **Heap Sort** | $O(n \log n)$ | $O(n \log n)$ | $O(n \log n)$ | $O(1)$ | No | Yes |
| **Timsort** | $O(n)$ | $O(n \log n)$ | $O(n \log n)$ | $O(n)$ | Yes | No |

## 2. Elementary $O(n^2)$ Sorts (Bubble, Selection, Insertion)


```python
# 1. Bubble Sort with Early Exit Optimization
def bubble_sort(arr):
    arr = list(arr)
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped: break  # Early exit if array is already sorted
    return arr

# 2. Selection Sort
def selection_sort(arr):
    arr = list(arr)
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

# 3. Insertion Sort
def insertion_sort(arr):
    arr = list(arr)
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

sample = [64, 34, 25, 12, 22, 11, 90]
print("Bubble Sort   :", bubble_sort(sample))
print("Selection Sort:", selection_sort(sample))
print("Insertion Sort:", insertion_sort(sample))
```

## 3. Advanced $O(n \log n)$ Sorts (Merge Sort, Quick Sort, Heap Sort)


```python
# 1. Merge Sort (Stable, O(n) space)
def merge_sort(arr):
    if len(arr) <= 1: return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    res = []; i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]: res.append(left[i]); i += 1
        else: res.append(right[j]); j += 1
    res.extend(left[i:]); res.extend(right[j:])
    return res

# 2. Quick Sort (In-Place, Lomuto Partitioning)
def quick_sort(arr):
    arr = list(arr)
    def _quick_sort(low, high):
        if low < high:
            p = partition(low, high)
            _quick_sort(low, p - 1)
            _quick_sort(p + 1, high)
    def partition(low, high):
        pivot = arr[high]
        i = low - 1
        for j in range(low, high):
            if arr[j] <= pivot:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]
        arr[i + 1], arr[high] = arr[high], arr[i + 1]
        return i + 1
    _quick_sort(0, len(arr) - 1)
    return arr

# 3. Heap Sort (O(1) auxiliary space)
import heapq
def heap_sort(arr):
    h = list(arr)
    heapq.heapify(h)
    return [heapq.heappop(h) for _ in range(len(h))]

print("Merge Sort:", merge_sort(sample))
print("Quick Sort:", quick_sort(sample))
print("Heap Sort :", heap_sort(sample))
```

## 4. Non-Comparison Linear Sorts (Counting Sort, Radix Sort)

Non-comparison sorts achieve **$O(n + k)$ time complexity** for integer range data.


```python
# Counting Sort for non-negative integers
def counting_sort(arr):
    if not arr: return arr
    max_val = max(arr)
    counts = [0] * (max_val + 1)
    for x in arr: counts[x] += 1
    res = []
    for val, count in enumerate(counts):
        res.extend([val] * count)
    return res

int_data = [4, 2, 2, 8, 3, 3, 1]
print("Counting Sort:", counting_sort(int_data))
```

## 5. Python's Built-in Timsort (`sorted()` & `list.sort()`)

**Timsort** (created by Tim Peters in 2002) is Python's default sorting algorithm.
- **Hybrid**: Combines **Adaptive Merge Sort** and **Insertion Sort**.
- **Runs**: Detects already-sorted contiguous sequences ("runs") in $O(n)$ best-case time.
- **Stable**: Preserves relative order of equal keys.


```python
students = [("Alice", 90), ("Bob", 75), ("Charlie", 90), ("Dave", 85)]

# Stable sort: secondary sort by name, primary sort by score
by_name = sorted(students, key=lambda s: s[0])
by_score = sorted(by_name, key=lambda s: s[1], reverse=True)

print("Stable Timsort result:")
for s in by_score: print(" ", s)
```

## 6. Empirical Performance Benchmarking


```python
import random, timeit
random.seed(42)

test_data = [random.randint(1, 1000) for _ in range(1000)]

t_bubble = timeit.timeit(lambda: bubble_sort(test_data), number=10)
t_merge  = timeit.timeit(lambda: merge_sort(test_data), number=10)
t_quick  = timeit.timeit(lambda: quick_sort(test_data), number=10)
t_tim    = timeit.timeit(lambda: sorted(test_data), number=10)

print("Execution time for 1000 items (10 runs):")
print(f"  Bubble Sort : {t_bubble:.4f}s")
print(f"  Merge Sort  : {t_merge:.4f}s")
print(f"  Quick Sort  : {t_quick:.4f}s")
print(f"  Timsort (C) : {t_tim:.6f}s (Built-in standard)")
```

## Summary

| Category | Algorithm | Best For |
|----------|-----------|----------|
| **Built-in** | **Timsort** (`sorted()`) | General-purpose sorting in Python ($O(n \log n)$ avg, $O(n)$ best) |

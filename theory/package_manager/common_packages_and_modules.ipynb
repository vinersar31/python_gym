## 1. Foundations: Modules vs. Packages

- **Module**: Any single `.py` file (e.g. `calculator.py`). It defines an isolated namespace.
- **Package**: A folder containing modules and an `__init__.py` file (e.g. `my_package/`).
- **Distribution Package**: A versioned archive (e.g. `.whl` or `.tar.gz`) published to PyPI containing one or more importable packages.


```python
import tempfile
import sys
from pathlib import Path

# Create a temporary custom module on disk
with tempfile.TemporaryDirectory() as tmp_dir:
    mod_path = Path(tmp_dir) / "math_helper.py"
    mod_path.write_text("""
def add(a: float, b: float) -> float:
    return a + b

def multiply(a: float, b: float) -> float:
    return a * b

PI = 3.14159265359
""", encoding="utf-8")

    # Add temp directory to sys.path
    sys.path.insert(0, tmp_dir)
    import math_helper

    print(f"Imported module: {math_helper.__name__}")
    print(f"math_helper.add(10, 25):      {math_helper.add(10, 25)}")
    print(f"math_helper.multiply(4, 7):   {math_helper.multiply(4, 7)}")
    print(f"math_helper.PI:               {math_helper.PI}")
    
    # Cleanup sys.path
    sys.path.remove(tmp_dir)
```

## 2. Import Mechanisms & Search Path (`sys.path`)

When you run `import foo`, Python searches the directories listed in `sys.path`:
1. Current working directory / script directory.
2. `PYTHONPATH` environment variable directories.
3. Standard Library directories.
4. Installed third-party `site-packages` directory.


```python
import sys

print("[Top 4 entries in sys.path search list]:")
for i, path_entry in enumerate(sys.path[:4]):
    print(f"  {i+1}. {path_entry}")
```

## 3. Dynamic Imports with `importlib`

Package managers, plugin engines, and CLI tools frequently load modules dynamically at runtime using `importlib.import_module()`.


```python
import importlib

# Dynamically import a module by string name
module_name = "math"
imported_math = importlib.import_module(module_name)

print(f"Dynamically imported {module_name}.sqrt(144): {imported_math.sqrt(144)}")
```

## 4. Module Execution & `if __name__ == '__main__':`

- When a file is run directly by the user: `__name__ == '__main__'`.
- When a file is imported as a library/package: `__name__ == 'module_name'`.


```python
print(f"Current namespace __name__ is: '{__name__}'")

def utility_task():
    return "Task completed successfully"

if __name__ == "__main__":
    print(f"Executed directly in main scope: {utility_task()}")
```

## 5. Essential Standard Library Modules ("Batteries Included")

Python bundles an extensive standard library that requires no external installation:

### 1. `pathlib.Path` & `json` (Modern File & Data Operations)


```python
from pathlib import Path
import json
import datetime

# 1. Path operations with pathlib.Path
current_file = Path.cwd()
print(f"Current Working Directory: {current_file}")

# 2. Data serialization with json
user_record = {
    "id": 42,
    "username": "ada_lovelace",
    "roles": ["engineer", "admin"],
    "created_at": str(datetime.date.today())
}

json_str = json.dumps(user_record, indent=2)
print("\nSerialized JSON:")
print(json_str)

# 3. Deserialization
restored = json.loads(json_str)
print(f"\nRestored username: {restored['username']} (id: {restored['id']})")
```

### 2. `collections`, `itertools`, & `functools` (High-Performance Utilities)


```python
from collections import Counter, defaultdict, deque
import itertools
import functools

# 1. Counter (Frequency analysis)
words = ["python", "java", "python", "rust", "python", "rust"]
freq = Counter(words)
print("Word frequencies:", freq.most_common(2))

# 2. defaultdict (Grouping without KeyErrors)
grouped = defaultdict(list)
items = [("fruit", "apple"), ("vegetable", "carrot"), ("fruit", "banana")]
for category, name in items:
    grouped[category].append(name)
print("Grouped items:", dict(grouped))

# 3. itertools.chain (Flattening multiple iterables)
combined = list(itertools.chain([1, 2], [3, 4], [5, 6]))
print("Chained list: ", combined)

# 4. functools.lru_cache (Memoization)
@functools.lru_cache(maxsize=128)
def fib(n: int) -> int:
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print("fib(30):      ", fib(30))
```

## 7. Package Architecture & Public API Control (`__all__`)

Using `__all__ = [...]` in a module or package `__init__.py` explicitly controls which symbols are exported when someone runs `from my_module import *`.


```python
# Simulating __all__ export control
__all__ = ["public_api_function"]

def public_api_function():
    return "I am public and exported!"

def _internal_private_helper():
    return "I am private!"

print(f"Defined __all__ export list: {__all__}")
```

## 8. Common Pitfalls & Best Practices

### ❌ Pitfall 1: Shadowing Standard Library Names
Creating a local file named `math.py`, `random.py`, or `json.py` will mask the Standard Library module, breaking any library that imports it.

### ❌ Pitfall 2: Circular Imports
If `module_a` imports `module_b` at the top level and `module_b` imports `module_a`, Python will raise an `ImportError` or `AttributeError`.
*Fix*: Move the import inside the specific function or refactor common shared logic into a third module `common.py`.

### ❌ Pitfall 3: Wildcard Imports (`from math import *`)
Pollutes the global namespace and causes hard-to-debug name collisions.

## 9. Hands-On Interactive Challenges


```python
import importlib
from collections import Counter
import heapq

# Challenge 1: Dynamic Driver / Plugin Dispatcher
def execute_dynamic_plugin(module_name: str, function_name: str, *args):
    """Dynamically loads a module and executes the specified function."""
    mod = importlib.import_module(module_name)
    func = getattr(mod, function_name)
    return func(*args)

# Challenge 2: Top-K Frequent Words using Counter and heapq
def get_top_k_words(text: str, k: int = 3) -> list[tuple[str, int]]:
    """Extracts the top-K most frequent words from a string."""
    words = [w.strip(".,!?\"'()").lower() for w in text.split() if w]
    counts = Counter(words)
    return heapq.nlargest(k, counts.items(), key=lambda item: item[1])

# Automated verification tests
# Test 1: Dynamic execution of math.pow
assert execute_dynamic_plugin("math", "pow", 2, 8) == 256.0
assert execute_dynamic_plugin("math", "floor", 9.7) == 9

# Test 2: Top-K words
sample_article = "Python is fast. Python is flexible. Python is everywhere. Learning Python is fun."
top_words = get_top_k_words(sample_article, k=2)
assert top_words[0] == ("python", 4)
assert top_words[1] == ("is", 4)

print("[OK] All Common Packages & Modules Challenges Passed!")
```

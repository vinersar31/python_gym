## 1. Introduction to OOP & The Class Paradigm

### 🧱 Procedural vs. Object-Oriented Programming

| Feature | Procedural Programming | Object-Oriented Programming (OOP) |
| :--- | :--- | :--- |
| **Primary Focus** | Step-by-step algorithms & functions | Modeling entities with state and behavior |
| **Data Organization** | Data and functions are separate | Data and functions are bundled together inside classes |
| **State Management** | State is passed through global or local variables | State is encapsulated within object instances |
| **Reusability** | Code reuse via functions and modules | Code reuse via inheritance, composition, and polymorphism |
| **Maintainability** | Can become hard to track as state grows | High modularity; changes to one class rarely break others |

### 🔍 "Everything in Python is an Object"
In Python, even primitive types (`int`, `str`, `float`), collections (`list`, `dict`), and functions are instances of built-in classes.


````python
# Everything in Python is an object!
num = 42
text = "Python"
numbers = [1, 2, 3]

def greet():
    return "Hello"

print(f"num={num} | type: {type(num)} | is object: {isinstance(num, object)}")
print(f"text='{text}' | type: {type(text)} | is object: {isinstance(text, object)}")
print(f"numbers={numbers} | type: {type(numbers)} | is object: {isinstance(numbers, object)}")
print(f"greet={greet.__name__} | type: {type(greet)} | is object: {isinstance(greet, object)}")

# Even types (classes) themselves are objects (instances of 'type')!
print(f"type(int): {type(int)} | type(list): {type(list)}")
print(f"isinstance(int, object): {isinstance(int, object)}")
````

### 🏗️ Blueprint vs. Instance Analogy
- A **Class** is the *architectural blueprint* (specifies what properties and actions exist).
- An **Instance (Object)** is the *actual physical house* built from that blueprint.
- You can build thousands of distinct houses from a single blueprint, each with its own wall color, furniture, and address.

## 3. The `__init__` Initializer & Demystifying `self`

### ⚙️ How `__init__` Works
- When you execute `user = User("Alice", "alice@example.com")`:
  1. Python calls `__new__()` to allocate the object in memory.
  2. Python immediately calls `__init__(self, "Alice", "alice@example.com")` to initialize the object's attributes.
  3. `__init__` must always return `None` (returning any value raises a `TypeError`).

### 🔑 Demystifying `self`
- `self` is a reference to the **specific instance** currently being operated on.
- In Python, `self` is an explicit parameter in method signatures, but Python supplies it automatically when you invoke `instance.method()`.
- Under the hood: `user.greet()` is converted into `User.greet(user)`!


````python
class Person:
    """Represents an individual person."""
    
    def __init__(self, name: str, age: int, city: str = "Unknown"):
        # Instance attributes bound to this specific instance
        self.name = name
        self.age = age
        self.city = city
        
    def introduce(self) -> str:
        """Instance method that reads attributes from self."""
        return f"Hi, I am {self.name}, {self.age} years old from {self.city}."
    
    def celebrate_birthday(self) -> None:
        """Instance method that mutates state on self."""
        self.age += 1
        print(f"[Happy Birthday] {self.name} is now {self.age} years old.")

# Create two distinct instances
p1 = Person("Alice", 28, "New York")
p2 = Person("Bob", 34, "San Francisco")

print(p1.introduce())
print(p2.introduce())

p1.celebrate_birthday()
print(f"p1 age: {p1.age} | p2 age: {p2.age} (p2 was unchanged!)")
````

### 🔬 Proof: `instance.method()` is Syntactic Sugar for `Class.method(instance)`


````python
# Standard invocation:
result1 = p1.introduce()

# Explicit Class-level invocation passing the instance manually:
result2 = Person.introduce(p1)

print(f"result1: {result1}")
print(f"result2: {result2}")
print(f"Are they identical? {result1 == result2}")
````

## 4. Instance Attributes vs. Class Attributes

### 📊 Key Differences

| Attribute Type | Where it is Defined | Scope & Storage | Memory Impact |
| :--- | :--- | :--- | :--- |
| **Instance Attribute** | Inside `__init__` or methods via `self.x` | Unique to each individual instance (`obj.__dict__`) | Stored separately for every instance |
| **Class Attribute** | Directly inside the `class` body | Shared across **all** instances of the class (`Class.__dict__`) | Stored once in the class object |

### 🧭 Attribute Lookup Resolution Order
When accessing `obj.attribute`:
1. Check instance dictionary: `obj.__dict__`
2. If not found, check class dictionary: `obj.__class__.__dict__`
3. If not found, search base classes in Method Resolution Order (MRO)
4. If still not found, raise `AttributeError` (or invoke `__getattr__` if defined)


````python
class Dog:
    # Class Attribute (shared by all dogs)
    species: str = "Canis lupus familiaris"
    total_dogs: int = 0

    def __init__(self, name: str, breed: str):
        # Instance Attributes (unique to each dog)
        self.name = name
        self.breed = breed
        
        # Modify the shared class attribute
        Dog.total_dogs += 1

# Create instances
d1 = Dog("Buddy", "Golden Retriever")
d2 = Dog("Luna", "German Shepherd")

print(f"d1 species: {d1.species} | d2 species: {d2.species}")
print(f"Dog.species: {Dog.species}")
print(f"Total dogs created: {Dog.total_dogs}")

# Inspecting namespaces (__dict__)
print(f"\nd1.__dict__: {d1.__dict__}")
print(f"d2.__dict__: {d2.__dict__}")
print(f"'species' in d1.__dict__: {'species' in d1.__dict__}")
print(f"'species' in Dog.__dict__: {'species' in Dog.__dict__}")
````

### ⚠️ The Shadowing Trap (Instance vs. Class Mutation)
Assigning to `self.species` creates a **new instance attribute** that *shadows* (hides) the class attribute for that instance only, without changing it for other instances!


````python
# Shadowing demonstration:
d1.species = "Wolf Hybrid"  # Creates 'species' key in d1.__dict__!

print(f"d1.species (shadowed): {d1.species}")
print(f"d2.species (still original class attr): {d2.species}")
print(f"Dog.species (class attr unchanged): {Dog.species}")
print(f"d1.__dict__: {d1.__dict__}")

# Deleting the instance attribute restores visibility of the class attribute!
del d1.species
print(f"After 'del d1.species': {d1.species}")
````

### 🚨 The Mutable Class Attribute Trap
If a class attribute is a **mutable** container (like a `list` or `dict`), modifying it in place via `self.attribute.append()` mutates the **shared class object** for *every* instance!


````python
# ANTI-PATTERN: Shared mutable class attribute
class BadStudent:
    enrolled_courses = []  # CLASS ATTRIBUTE: Shared list!

    def __init__(self, name: str):
        self.name = name

    def add_course(self, course: str):
        self.enrolled_courses.append(course)  # Modifies shared class list!

s1 = BadStudent("Alice")
s2 = BadStudent("Bob")

s1.add_course("Python 101")
print(f"s1 courses: {s1.enrolled_courses}")
print(f"s2 courses: {s2.enrolled_courses}  <-- Bob has Alice's course!")

# CORRECT PATTERN: Initialize mutable containers in __init__
class GoodStudent:
    def __init__(self, name: str):
        self.name = name
        self.enrolled_courses = []  # INSTANCE ATTRIBUTE: Fresh list per student

    def add_course(self, course: str):
        self.enrolled_courses.append(course)

g1 = GoodStudent("Alice")
g2 = GoodStudent("Bob")
g1.add_course("Python 101")

print(f"\ng1 courses: {g1.enrolled_courses}")
print(f"g2 courses: {g2.enrolled_courses}  <-- Bob's list is completely independent!")
````

## 5. Method Types: Instance, Class, & Static Methods

Python supports three distinct types of methods inside a class:

| Method Type | Decorator | 1st Argument | Can Access/Modify | Primary Use Cases |
| :--- | :--- | :--- | :--- | :--- |
| **Instance Method** | *None* | `self` | Instance State (`self`) & Class State (`type(self)`) | Normal behavior that operates on an instance's state |
| **Class Method** | `@classmethod` | `cls` | Class State (`cls`) only | Factory constructors (`from_dict`, `from_csv`), class-wide state tracking |
| **Static Method** | `@staticmethod` | *None* | Neither `self` nor `cls` (Isolated) | Utility / helper functions logically grouped inside the class |


````python
from datetime import date

class Employee:
    # Class attribute
    company_name: str = "TechCorp Global"
    minimum_wage: float = 30000.0

    def __init__(self, name: str, salary: float, birth_year: int):
        self.name = name
        self.salary = max(salary, self.minimum_wage)
        self.birth_year = birth_year

    # 1. Instance Method
    def get_info(self) -> str:
        return f"{self.name} works at {self.company_name} earning ${self.salary:,.2f}"

    def give_raise(self, percentage: float) -> None:
        self.salary *= (1 + percentage / 100)

    # 2. Class Method: Alternative Constructor (Factory Pattern)
    @classmethod
    def from_birth_year(cls, name: str, salary: float, birth_year: int):
        """Creates an employee instance directly."""
        return cls(name, salary, birth_year)

    @classmethod
    def from_string(cls, employee_str: str):
        """Factory method: parses 'Name,Salary,BirthYear' formatted strings."""
        parts = employee_str.split(",")
        name = parts[0].strip()
        salary = float(parts[1].strip())
        birth_year = int(parts[2].strip())
        return cls(name, salary, birth_year)

    @classmethod
    def set_company_name(cls, new_name: str) -> None:
        """Class method that updates class-level configuration."""
        cls.company_name = new_name

    # 3. Static Method: Utility function with no access to self or cls
    @staticmethod
    def calculate_age(birth_year: int, current_year: int = None) -> int:
        """Pure utility function to compute age."""
        if current_year is None:
            current_year = date.today().year
        return current_year - birth_year

    @staticmethod
    def is_valid_salary(salary: float) -> bool:
        """Validates if a given salary meets minimum threshold."""
        return salary >= Employee.minimum_wage

# Usage:
emp1 = Employee("Sarah Connor", 75000.0, 1990)
print(emp1.get_info())

# Using @classmethod factory constructor:
emp2 = Employee.from_string("John Matrix, 92000.0, 1985")
print(emp2.get_info())

# Using @staticmethod utility:
age = Employee.calculate_age(emp1.birth_year)
print(f"{emp1.name} is {age} years old.")
print(f"Is $25,000 valid salary? {Employee.is_valid_salary(25000)}")

# Changing class state affects all instances
Employee.set_company_name("Cyberdyne Systems")
print(f"\nAfter company rename:")
print(emp1.get_info())
print(emp2.get_info())
````

## 6. Object Lifecycle, Identity, & Introspection

### 🔄 The Complete Object Lifecycle

```text
1. Instantiation: Class(*args)
      │
      ▼
2. Allocation: __new__(cls, *args)   --> Allocates memory & returns new instance
      │
      ▼
3. Initialization: __init__(self, *args) --> Sets up attributes on self
      │
      ▼
4. Active Usage: Method calls, attribute access, operations
      │
      ▼
5. Finalization: Ref count reaches 0 --> __del__(self) finalizer (Garbage Collected)
```


````python
class LifecycleTracker:
    def __new__(cls, *args, **kwargs):
        print("  [1. __new__] Allocating raw memory for instance...")
        instance = super().__new__(cls)
        return instance

    def __init__(self, label: str):
        print(f"  [2. __init__] Initializing instance with label='{label}'...")
        self.label = label

    def __del__(self):
        print(f"  [3. __del__] Finalizing & destroying instance '{self.label}'!")

print("Creating object:")
tracker = LifecycleTracker("DemoObj")

print("\nDeleting object reference:")
del tracker
````

### 🔍 Introspection Toolkit
Python provides rich built-in functions to inspect any object at runtime:

- `isinstance(obj, Class)`: Checks if `obj` is an instance of `Class` (or a subclass).
- `issubclass(Sub, Super)`: Checks class hierarchy.
- `hasattr(obj, 'attr')`: Returns `True` if attribute exists.
- `getattr(obj, 'attr', default)`: Retrieves attribute value safely.
- `setattr(obj, 'attr', value)`: Sets attribute dynamically.
- `delattr(obj, 'attr')`: Deletes attribute dynamically.
- `vars(obj)`: Returns `obj.__dict__`.
- `dir(obj)`: Returns list of all valid attributes and methods.


````python
class DynamicUser:
    def __init__(self, username: str, email: str):
        self.username = username
        self.email = email

user = DynamicUser("alex99", "alex@example.com")

# Dynamic attribute operations
print("hasattr 'email':", hasattr(user, "email"))
print("hasattr 'role':", hasattr(user, "role"))

# getattr with fallback
role = getattr(user, "role", "guest")
print(f"Retrieved role: {role}")

# setattr creates attribute on the fly
setattr(user, "role", "admin")
print(f"After setattr: user.role = {user.role}")

# vars() vs dir()
print(f"\nvars(user): {vars(user)}")
print(f"dir(user) (sample): {[a for a in dir(user) if not a.startswith('__')]}")
````

## 7. Memory Optimization with `__slots__`

### 💡 Why `__slots__`?
- By default, Python stores instance attributes in a dynamic dictionary (`self.__dict__`).
- While flexible, dictionaries consume significant memory overhead per instance.
- If your application instantiates **hundreds of thousands or millions** of small objects (e.g., coordinates, nodes, rows), this overhead becomes huge.
- By defining `__slots__`, Python uses a fixed-size compact array instead of `__dict__`, saving up to **60–70% of memory** and boosting attribute access speed!


````python
import sys
import time

# Standard class with __dict__
class StandardPoint:
    def __init__(self, x: float, y: float, z: float):
        self.x = x
        self.y = y
        self.z = z

# Optimized class with __slots__
class SlottedPoint:
    __slots__ = ('x', 'y', 'z')
    
    def __init__(self, x: float, y: float, z: float):
        self.x = x
        self.y = y
        self.z = z

p_std = StandardPoint(1.0, 2.0, 3.0)
p_slot = SlottedPoint(1.0, 2.0, 3.0)

print(f"StandardPoint instance base size: {sys.getsizeof(p_std)} bytes")
print(f"StandardPoint __dict__ size:      {sys.getsizeof(p_std.__dict__)} bytes")
print(f"Total StandardPoint footprint:    {sys.getsizeof(p_std) + sys.getsizeof(p_std.__dict__)} bytes")

print(f"\nSlottedPoint instance size:       {sys.getsizeof(p_slot)} bytes (No __dict__!)")

# Demonstrate that dynamic attribute assignment is prevented
try:
    p_slot.w = 4.0
except AttributeError as e:
    print(f"\n__slots__ prevents arbitrary attributes: {e}")
````

````python
# Memory & Performance Benchmark: Creating 100,000 instances
N = 100_000

t0 = time.perf_counter()
std_list = [StandardPoint(float(i), float(i+1), float(i+2)) for i in range(N)]
t_std = time.perf_counter() - t0

t0 = time.perf_counter()
slot_list = [SlottedPoint(float(i), float(i+1), float(i+2)) for i in range(N)]
t_slot = time.perf_counter() - t0

print(f"Time to create {N:,} StandardPoint instances: {t_std*1000:.2f} ms")
print(f"Time to create {N:,} SlottedPoint instances:  {t_slot*1000:.2f} ms")
print(f"Speedup: {t_std / t_slot:.2f}x faster with __slots__")
````

### 8.1 Bank Account Management System
A robust financial account modeling balances, deposits, withdrawals, transactions, and interest calculations.


````python
from datetime import datetime

class BankAccount:
    """Models a commercial bank account with full ledger tracking."""
    
    # Class attributes
    bank_name: str = "Apex Federal Bank"
    routing_number: str = "021000021"
    base_interest_rate: float = 0.035  # 3.5%
    _account_sequence: int = 100000

    def __init__(self, owner: str, initial_deposit: float = 0.0):
        if initial_deposit < 0:
            raise ValueError("Initial deposit cannot be negative.")
        
        BankAccount._account_sequence += 1
        self.account_number = str(BankAccount._account_sequence)
        self.owner = owner
        self._balance = float(initial_deposit)
        self._transactions: list[dict] = []
        
        self._log_transaction("INITIAL_DEPOSIT", initial_deposit)

    @property
    def balance(self) -> float:
        """Read-only balance property."""
        return self._balance

    def deposit(self, amount: float) -> float:
        """Deposits funds into the account."""
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self._balance += amount
        self._log_transaction("DEPOSIT", amount)
        return self._balance

    def withdraw(self, amount: float) -> float:
        """Withdraws funds if sufficient balance exists."""
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self._balance:
            raise ValueError(f"Insufficient funds. Balance: ${self._balance:,.2f}, Requested: ${amount:,.2f}")
        
        self._balance -= amount
        self._log_transaction("WITHDRAWAL", -amount)
        return self._balance

    def apply_interest(self) -> float:
        """Applies annual interest based on class interest rate."""
        interest = self._balance * self.base_interest_rate
        self._balance += interest
        self._log_transaction("INTEREST", interest)
        return interest

    def _log_transaction(self, tx_type: str, amount: float) -> None:
        self._transactions.append({
            "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
            "type": tx_type,
            "amount": amount,
            "resulting_balance": self._balance
        })

    def get_statement(self) -> str:
        """Formats a printed account ledger statement."""
        lines = [
            "=" * 50,
            f"{self.bank_name} - ACCOUNT STATEMENT",
            f"Account #: {self.account_number} | Owner: {self.owner}",
            f"Current Balance: ${self._balance:,.2f}",
            "-" * 50,
            f"{'Type':<18} | {'Amount ($)':<12} | {'Balance ($)':<12}",
            "-" * 50
        ]
        for tx in self._transactions:
            lines.append(f"{tx['type']:<18} | {tx['amount']:>12,.2f} | {tx['resulting_balance']:>12,.2f}")
        lines.append("=" * 50)
        return "\n".join(lines)

    @classmethod
    def set_interest_rate(cls, new_rate: float) -> None:
        if new_rate < 0:
            raise ValueError("Interest rate cannot be negative.")
        cls.base_interest_rate = new_rate

# Demo BankAccount
acc = BankAccount("Ada Lovelace", 5000.0)
acc.deposit(1250.0)
acc.withdraw(800.0)
acc.apply_interest()
print(acc.get_statement())
````

### 8.2 2D Vector Physics Engine
Modeling mathematical coordinates, Euclidean distance, magnitude, scalar operations, and static angle helpers.


````python
import math

class Vector2D:
    """A 2D geometric vector with physics computations."""
    __slots__ = ('x', 'y')

    def __init__(self, x: float, y: float):
        self.x = float(x)
        self.y = float(y)

    def magnitude(self) -> float:
        """Returns the Euclidean length (magnitude) of the vector."""
        return math.hypot(self.x, self.y)

    def dot(self, other: 'Vector2D') -> float:
        """Computes the scalar dot product with another vector."""
        return self.x * other.x + self.y * other.y

    def scale(self, factor: float) -> 'Vector2D':
        """Returns a new scaled Vector2D."""
        return Vector2D(self.x * factor, self.y * factor)

    def normalized(self) -> 'Vector2D':
        """Returns a unit vector in the same direction."""
        mag = self.magnitude()
        if mag == 0:
            raise ValueError("Cannot normalize a zero-length vector.")
        return Vector2D(self.x / mag, self.y / mag)

    @classmethod
    def from_polar(cls, radius: float, angle_radians: float) -> 'Vector2D':
        """Alternative constructor: creates a vector from polar coordinates."""
        return cls(radius * math.cos(angle_radians), radius * math.sin(angle_radians))

    @staticmethod
    def distance(v1: 'Vector2D', v2: 'Vector2D') -> float:
        """Static utility to compute Euclidean distance between two vector positions."""
        return math.hypot(v1.x - v2.x, v1.y - v2.y)

    def __repr__(self) -> str:
        return f"Vector2D(x={self.x:.2f}, y={self.y:.2f})"

# Demo Vector2D
v1 = Vector2D(3, 4)
v2 = Vector2D(6, 8)

print(f"v1: {v1} | magnitude: {v1.magnitude()}")
print(f"v1 normalized: {v1.normalized()}")
print(f"v1 dot v2: {v1.dot(v2)}")
print(f"Distance between v1 and v2: {Vector2D.distance(v1, v2)}")

# From polar coordinates (radius=10, angle=45 degrees)
v_polar = Vector2D.from_polar(10, math.radians(45))
print(f"v_polar: {v_polar} | magnitude: {v_polar.magnitude():.2f}")
````

### 8.3 E-Commerce Product & Inventory Catalog
Demonstrates class-level instance registries, category grouping, and JSON/dict alternative constructors.


````python
class Product:
    """Catalog product with central registry and category tracking."""
    
    # Class registry tracking all active products
    _registry: dict[str, 'Product'] = {}
    
    def __init__(self, sku: str, name: str, price: float, category: str, stock: int = 0):
        sku = sku.upper().strip()
        if sku in Product._registry:
            raise ValueError(f"Product with SKU '{sku}' already exists in registry.")
        if price < 0 or stock < 0:
            raise ValueError("Price and stock cannot be negative.")
            
        self.sku = sku
        self.name = name
        self.price = float(price)
        self.category = category.title().strip()
        self.stock = int(stock)
        
        # Register in class-level registry
        Product._registry[self.sku] = self

    def apply_discount(self, percent: float) -> None:
        if not (0 <= percent <= 100):
            raise ValueError("Discount must be between 0 and 100%.")
        self.price *= (1 - percent / 100)

    def restock(self, quantity: int) -> None:
        if quantity <= 0:
            raise ValueError("Restock quantity must be positive.")
        self.stock += quantity

    @classmethod
    def from_dict(cls, data: dict) -> 'Product':
        """Factory method to instantiate a Product from a dictionary / JSON payload."""
        return cls(
            sku=data["sku"],
            name=data["name"],
            price=data["price"],
            category=data["category"],
            stock=data.get("stock", 0)
        )

    @classmethod
    def get_by_sku(cls, sku: str) -> 'Product | None':
        return cls._registry.get(sku.upper().strip())

    @classmethod
    def get_total_inventory_value(cls) -> float:
        """Calculates total value across all registered products."""
        return sum(p.price * p.stock for p in cls._registry.values())

    @classmethod
    def list_by_category(cls, category: str) -> list['Product']:
        category = category.title().strip()
        return [p for p in cls._registry.values() if p.category == category]

    def __repr__(self) -> str:
        return f"<Product [{self.sku}] {self.name} - ${self.price:.2f} (Stock: {self.stock})>"

# Populate catalog
Product._registry.clear()  # reset for notebook rerun

p1 = Product("LAP-001", "MacBook Pro 16", 2499.00, "Electronics", stock=15)
p2 = Product("MOU-002", "Logitech MX Master 3S", 99.99, "Electronics", stock=50)
p3 = Product.from_dict({
    "sku": "DESK-003",
    "name": "Standing Desk Pro",
    "price": 499.00,
    "category": "Furniture",
    "stock": 8
})

print(f"Total products registered: {len(Product._registry)}")
print(f"Total inventory valuation: ${Product.get_total_inventory_value():,.2f}")
print(f"Electronics products: {Product.list_by_category('Electronics')}")
````

## 9. Common Pitfalls & Anti-Patterns

### ❌ Pitfall 1: Mutable Default Arguments in `__init__`
Never use mutable objects (`list`, `dict`, `set`) as default values in `__init__`. The default is evaluated **once when the class is parsed**, so all instances sharing the default will point to the exact same list!


````python
# BAD:
class BadCart:
    def __init__(self, owner: str, items: list = []):  # SHARED DEFAULT LIST!
        self.owner = owner
        self.items = items

# GOOD:
class GoodCart:
    def __init__(self, owner: str, items: list = None):
        self.owner = owner
        self.items = items if items is not None else []  # Fresh list per instance

cart1 = BadCart("Alice")
cart1.items.append("Apple")
cart2 = BadCart("Bob")
print(f"BadCart cart2 items: {cart2.items}  <-- Polluted with Alice's Apple!")

g_cart1 = GoodCart("Alice")
g_cart1.items.append("Apple")
g_cart2 = GoodCart("Bob")
print(f"GoodCart g_cart2 items: {g_cart2.items} <-- Clean and empty!")
````

### ❌ Pitfall 2: Overwriting Class State Unintentionally
Using `self.__class__.attribute = value` modifies the shared state for every instance of that class. Use this only when intentional.


### ❌ Pitfall 3: Relying on `__del__` for Deterministic Cleanup
Python's garbage collector does not guarantee *when* or even *if* `__del__` will run (e.g. during program exit or circular references). Always use **Context Managers** (`with` statements) for deterministic cleanup (files, sockets, database locks).

### 🎯 Challenge 1: Smart Thermostat Class
Create a class `Thermostat` with:
- Instance attributes: `location: str`, `target_temp: float` (default 20.0), `mode: str` (default "OFF").
- Class attributes: `MIN_TEMP = 10.0`, `MAX_TEMP = 35.0`, `VALID_MODES = {"OFF", "HEAT", "COOL"}`.
- Methods:
  - `set_target_temp(temp: float)`: Updates temperature within `[MIN_TEMP, MAX_TEMP]`, raises `ValueError` if out of bounds.
  - `set_mode(mode: str)`: Updates mode (case-insensitive) if in `VALID_MODES`, raises `ValueError` otherwise.
  - `@staticmethod c_to_f(celsius: float) -> float`: Converts Celsius to Fahrenheit ($F = C \times \frac{9}{5} + 32$).


````python
class Thermostat:
    MIN_TEMP: float = 10.0
    MAX_TEMP: float = 35.0
    VALID_MODES: set[str] = {"OFF", "HEAT", "COOL"}

    def __init__(self, location: str, target_temp: float = 20.0, mode: str = "OFF"):
        self.location = location
        self._validate_and_set_temp(target_temp)
        self.set_mode(mode)

    def _validate_and_set_temp(self, temp: float) -> None:
        if not (self.MIN_TEMP <= temp <= self.MAX_TEMP):
            raise ValueError(f"Temperature {temp}°C out of range [{self.MIN_TEMP}, {self.MAX_TEMP}]")
        self.target_temp = float(temp)

    def set_target_temp(self, temp: float) -> None:
        self._validate_and_set_temp(temp)

    def set_mode(self, mode: str) -> None:
        mode_upper = mode.upper().strip()
        if mode_upper not in self.VALID_MODES:
            raise ValueError(f"Invalid mode '{mode}'. Valid modes: {self.VALID_MODES}")
        self.mode = mode_upper

    @staticmethod
    def c_to_f(celsius: float) -> float:
        return celsius * 9 / 5 + 32

# Automated verification test
t = Thermostat("Living Room", 22.5, "heat")
assert t.location == "Living Room"
assert t.target_temp == 22.5
assert t.mode == "HEAT"
assert Thermostat.c_to_f(0) == 32.0
assert Thermostat.c_to_f(100) == 212.0

# Boundary testing
try:
    t.set_target_temp(40.0)
    assert False, "Should have raised ValueError for high temp"
except ValueError:
    pass

try:
    t.set_mode("TURBO")
    assert False, "Should have raised ValueError for invalid mode"
except ValueError:
    pass

print("[OK] Challenge 1 Passed!")
````

### 🎯 Challenge 2: Employee Payroll Pool with Factory Constructor
Implement an `EmployeePayroll` class:
- Class attributes: `total_payroll_pool: float = 0.0`, `employee_count: int = 0`.
- Instance attributes: `emp_id: str`, `name: str`, `salary: float`.
- Factory constructor: `@classmethod from_csv_row(cls, csv_text: str)` parsing `"E101,Jane Doe,85000"`.
- Method `give_bonus(amount: float)` that increases instance salary and updates the global `total_payroll_pool`.


````python
class EmployeePayroll:
    total_payroll_pool: float = 0.0
    employee_count: int = 0

    def __init__(self, emp_id: str, name: str, salary: float):
        self.emp_id = emp_id.strip()
        self.name = name.strip()
        self.salary = float(salary)
        
        # Update class pool
        EmployeePayroll.employee_count += 1
        EmployeePayroll.total_payroll_pool += self.salary

    def give_bonus(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError("Bonus must be positive.")
        self.salary += amount
        EmployeePayroll.total_payroll_pool += amount

    @classmethod
    def from_csv_row(cls, csv_text: str) -> 'EmployeePayroll':
        emp_id, name, salary_str = csv_text.split(",")
        return cls(emp_id, name, float(salary_str))

    @classmethod
    def reset_metrics(cls) -> None:
        cls.total_payroll_pool = 0.0
        cls.employee_count = 0

# Automated verification test
EmployeePayroll.reset_metrics()

e1 = EmployeePayroll("E101", "Alice", 60000.0)
e2 = EmployeePayroll.from_csv_row("E102, Bob, 80000.0")

assert EmployeePayroll.employee_count == 2
assert EmployeePayroll.total_payroll_pool == 140000.0

e1.give_bonus(5000.0)
assert e1.salary == 65000.0
assert EmployeePayroll.total_payroll_pool == 145000.0

print("[OK] Challenge 2 Passed!")
````

### 🎯 Challenge 3: In-Memory Task State Machine
Implement a `Task` class:
- Allowed transitions: `PENDING` $
ightarrow$ `RUNNING` $
ightarrow$ `COMPLETED` or `FAILED`.
- Class attribute `tasks_by_status: dict[str, list]` grouping tasks.
- Methods: `start()`, `complete()`, `fail()` enforcing valid transitions.


````python
class Task:
    STATUS_PENDING = "PENDING"
    STATUS_RUNNING = "RUNNING"
    STATUS_COMPLETED = "COMPLETED"
    STATUS_FAILED = "FAILED"

    def __init__(self, title: str):
        self.title = title
        self.status = self.STATUS_PENDING

    def start(self) -> None:
        if self.status != self.STATUS_PENDING:
            raise ValueError(f"Cannot start task in '{self.status}' state.")
        self.status = self.STATUS_RUNNING

    def complete(self) -> None:
        if self.status != self.STATUS_RUNNING:
            raise ValueError(f"Cannot complete task from '{self.status}' state.")
        self.status = self.STATUS_COMPLETED

    def fail(self) -> None:
        if self.status != self.STATUS_RUNNING:
            raise ValueError(f"Cannot fail task from '{self.status}' state.")
        self.status = self.STATUS_FAILED

# Automated verification test
t = Task("Database Migration")
assert t.status == Task.STATUS_PENDING

t.start()
assert t.status == Task.STATUS_RUNNING

t.complete()
assert t.status == Task.STATUS_COMPLETED

try:
    t.start()  # Invalid transition from COMPLETED
    assert False, "Should raise ValueError"
except ValueError:
    pass

print("[OK] Challenge 3 Passed!")
````

## 11. Quick Reference Card & Summary

### 💡 All Core Syntax in One Runnable Cell


````python
# ============================================================
# PYTHON CLASSES & OBJECTS — QUICK REFERENCE CARD
# ============================================================

class QuickRefAccount:
    # 1. Class attribute (shared by all instances)
    account_type: str = "Standard"
    total_accounts: int = 0

    # 2. Memory optimization (optional)
    # __slots__ = ('owner', 'balance')

    # 3. Initializer
    def __init__(self, owner: str, balance: float = 0.0):
        self.owner = owner          # Instance attribute
        self.balance = balance      # Instance attribute
        QuickRefAccount.total_accounts += 1

    # 4. Instance Method (receives `self`)
    def deposit(self, amount: float) -> float:
        self.balance += amount
        return self.balance

    # 5. Class Method (receives `cls`, alternative constructor)
    @classmethod
    def from_csv(cls, data_str: str) -> 'QuickRefAccount':
        owner, balance = data_str.split(",")
        return cls(owner.strip(), float(balance.strip()))

    # 6. Static Method (utility function, no self/cls)
    @staticmethod
    def is_valid_currency(symbol: str) -> bool:
        return symbol in {"USD", "EUR", "GBP", "JPY"}

    # 7. Representation
    def __repr__(self) -> str:
        return f"QuickRefAccount(owner='{self.owner}', balance={self.balance})"

# Quick demo
acc = QuickRefAccount.from_csv("Linus Torvalds, 10000.0")
acc.deposit(2500.0)
print(f"Created: {acc}")
print(f"Total accounts: {QuickRefAccount.total_accounts}")
print(f"Is USD valid currency? {QuickRefAccount.is_valid_currency('USD')}")
print(f"Is instance of QuickRefAccount: {isinstance(acc, QuickRefAccount)}")
````

### 📊 Summary Cheat Sheet

| Concept | Syntax / Mechanism | Purpose / Key Takeaway |
| :--- | :--- | :--- |
| **Class Definition** | `class Name:` | Blueprint declaring structure and behavior (PascalCase) |
| **Instantiation** | `obj = Name(*args)` | Creates new instance in memory |
| **Initializer** | `def __init__(self, ...):` | Initializes instance attributes; returns `None` |
| **Instance Reference** | `self` | Explicit reference to current object instance |
| **Instance Attribute** | `self.attr = val` | Stored in `obj.__dict__`, unique per instance |
| **Class Attribute** | `attr = val` (class body) | Stored in `Class.__dict__`, shared by all instances |
| **Instance Method** | `def foo(self):` | Operates on instance state |
| **Class Method** | `@classmethod def foo(cls):` | Operates on class state / factory constructors |
| **Static Method** | `@staticmethod def foo():` | Pure helper function namespaced inside class |
| **Memory Slots** | `__slots__ = (...)` | Eliminates `__dict__` overhead, saves ~60% RAM |
| **Introspection** | `isinstance`, `hasattr`, `getattr` | Dynamic runtime object examination |

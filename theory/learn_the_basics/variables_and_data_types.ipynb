## 1. Variables

A **variable** is a named location in memory used to store a value.
In Python you do **not** declare a type — a variable is created the moment
you assign a value to it with `=`.

```
variable_name = value
```

### Rules for variable names
- Start with a letter (`a-z`, `A-Z`) or underscore `_`
- Followed by letters, digits, or underscores
- Case-sensitive (`age` ≠ `Age`)
- Cannot be a Python keyword


````python
# Basic variable assignment
name    = "Alice"       # str
age     = 30            # int
height  = 1.75          # float
is_student = True       # bool

print(f"Name   : {name}")
print(f"Age    : {age}")
print(f"Height : {height} m")
print(f"Student: {is_student}")
````

````python
# Variables are just labels pointing to objects in memory
x = 42
print(f"x = {x}, id(x) = {id(x)}")

x = 99          # reassign – now x points to a different object
print(f"x = {x}, id(x) = {id(x)}")
````

## 2. Multiple Assignment

Python offers several convenient shorthand forms for assigning values.


````python
# Assign the same value to multiple variables at once
a = b = c = 0
print(f"a={a}, b={b}, c={c}")
````

````python
# Assign different values to multiple variables in one line (unpacking)
x, y, z = "apple", 10, 3.14
print(f"x={x}, y={y}, z={z}")
````

````python
# Swap two variables without a temporary variable
a, b = 5, 10
print(f"Before swap: a={a}, b={b}")
a, b = b, a
print(f"After swap : a={a}, b={b}")
````

````python
# Extended unpacking with *rest
first, *rest = [1, 2, 3, 4, 5]
*beginning, last = [10, 20, 30, 40]
print(f"first={first}, rest={rest}")
print(f"beginning={beginning}, last={last}")
````

````python
# Augmented assignment operators
n = 10
n += 5;  print(f'+= : {n}')  # 15
n -= 3;  print(f'-= : {n}')  # 12
n *= 2;  print(f'*= : {n}')  # 24
n //= 5; print(f'//=: {n}')  # 4
n **= 3; print(f'**=: {n}')  # 64
n %= 10; print(f'%=  : {n}') # 4
````

## 3. Dynamic Typing

Python is **dynamically typed**: the type of a variable is determined at
**runtime**, not at declaration time. The same variable can hold different
types throughout its lifetime.

| Feature | Statically Typed (C/Java) | Dynamically Typed (Python) |
|---------|--------------------------|----------------------------|
| Type declared? | Yes (`int x = 5;`) | No (`x = 5`) |
| Type checked? | Compile time | Runtime |
| Can reassign type? | No | Yes |


````python
var = 42
print(f"var = {var!r}  type: {type(var).__name__}")

var = "hello"
print(f"var = {var!r}  type: {type(var).__name__}")

var = [1, 2, 3]
print(f"var = {var!r}  type: {type(var).__name__}")

var = None
print(f"var = {var!r}  type: {type(var).__name__}")
````

## 4. Data Types Overview

Python has the following **built-in data types**:

| Category | Type(s) | Example literal |
|----------|---------|-----------------|
| **Text** | `str` | `"hello"` |
| **Numeric** | `int`, `float`, `complex` | `42`, `3.14`, `2+3j` |

## 5. Numbers

Python has **three numeric types**:

| Type | Description | Example |
|------|-------------|---------|
| `int` | Arbitrary-precision integer | `42`, `-7`, `10_000_000` |
| `float` | 64-bit floating point (IEEE 754) | `3.14`, `-0.5`, `1e-3` |
| `complex` | Real + imaginary part | `2+3j`, `complex(1, -2)` |


````python
# --- int ---
a = 42
b = -7
big = 10_000_000   # underscores for readability
binary = 0b1010    # binary literal
octal  = 0o17      # octal literal
hexa   = 0xFF      # hexadecimal literal

print(f"int    : a={a}, b={b}, big={big}")
print(f"bases  : binary={binary}, octal={octal}, hex={hexa}")
print(f"Python int has no max: {2**1000}")
````

````python
# --- float ---
pi    = 3.14159
sci   = 1.5e-3     # scientific notation
neg   = -0.001

print(f"float  : pi={pi}, sci={sci}, neg={neg}")

# Floating-point precision gotcha
print(f"0.1 + 0.2 = {0.1 + 0.2}")          # not exactly 0.3!

import math
print(f"math.inf = {math.inf}")
print(f"math.nan = {math.nan}")
````

````python
# --- complex ---
z1 = 2 + 3j
z2 = complex(1, -2)

print(f"z1          = {z1}")
print(f"z2          = {z2}")
print(f"z1 + z2     = {z1 + z2}")
print(f"z1.real     = {z1.real}")
print(f"z1.imag     = {z1.imag}")
print(f"abs(z1)     = {abs(z1):.4f}")
````

````python
# Arithmetic operators
x, y = 17, 5
print(f"{x} + {y}  = {x + y}")
print(f"{x} - {y}  = {x - y}")
print(f"{x} * {y}  = {x * y}")
print(f"{x} / {y}  = {x / y}")   # true division (float)
print(f"{x} // {y} = {x // y}")  # floor division (int)
print(f"{x} % {y}  = {x % y}")   # modulo
print(f"{x} ** {y} = {x ** y}")  # exponentiation
````

## 6. Strings

A **string** (`str`) is an immutable sequence of Unicode characters.
Strings support a rich set of operations and methods.

| Creation | Example |
|----------|---------|
| Single quotes | `'hello'` |
| Double quotes | `"hello"` |
| Triple quotes | `"""multi-line"""` |
| Raw string | `r"C:\Users\name"` |
| f-string | `f"Hello, {name}"` |


````python
# String creation
s1 = 'single quotes'
s2 = "double quotes"
s3 = """triple\nquoted"""
s4 = r"C:\Users\Alice\Desktop"   # raw – backslashes are literal
s5 = f"2 + 2 = {2 + 2}"          # f-string (Python 3.6+)

for s in [s1, s2, s3, s4, s5]:
    print(repr(s))
````

````python
# String indexing & slicing
word = "Python"
#       P  y  t  h  o  n
#       0  1  2  3  4  5
#      -6 -5 -4 -3 -2 -1

print(f"word[0]   = {word[0]}")    # first char
print(f"word[-1]  = {word[-1]}")   # last char
print(f"word[1:4] = {word[1:4]}")  # slice [start:stop]
print(f"word[::-1]= {word[::-1]}") # reverse
print(f"len(word) = {len(word)}")
````

````python
# Common string methods
msg = "  Hello, World!  "
print(msg.strip())          # remove leading/trailing whitespace
print(msg.lower())
print(msg.upper())
print(msg.replace("World", "Python"))
print(msg.split(","))       # returns a list

# Checking content
print("Python".startswith("Py"))  # True
print("hello123".isalnum())       # True
print("  	
".isspace())         # True

# Joining
words = ["join", "these", "words"]
print(" | ".join(words))
````

````python
# f-strings (formatted string literals)
name   = "Alice"
score  = 98.5
passed = True

print(f"Student : {name}")
print(f"Score   : {score:.1f}%")
print(f"Passed  : {passed}")
print(f"Upper   : {name.upper()}")
print(f"Expr    : {2 ** 10 = }")  # Python 3.8+ self-documenting
````

````python
# String immutability
greeting = "Hello"
# greeting[0] = 'J'   # TypeError: 'str' object does not support item assignment

# To 'modify', create a new string
new_greeting = "J" + greeting[1:]
print(new_greeting)
````

## 7. Booleans

The `bool` type has exactly **two values**: `True` and `False`.
It is a subclass of `int` — `True == 1` and `False == 0`.


````python
# Boolean literals
t = True
f = False
print(type(t), t, f)

# bool is a subclass of int
print(f"True  + True  = {True  + True}")
print(f"True  + False = {True  + False}")
print(f"True  * 10    = {True  * 10}")
````

````python
# Boolean operators: and, or, not
a, b = True, False
print(f"a and b = {a and b}")
print(f"a or  b = {a or  b}")
print(f"not a   = {not a}")
````

````python
# Comparison operators return bool
x = 10
print(f"x == 10  : {x == 10}")
print(f"x != 5   : {x != 5}")
print(f"x > 5    : {x > 5}")
print(f"x <= 10  : {x <= 10}")
print(f"5 < x <= 20: {5 < x <= 20}")   # chained comparison
````

````python
# Truthy vs Falsy values
falsy_values = [False, 0, 0.0, 0j, '', [], (), {}, set(), None]

print("Falsy values:")
for v in falsy_values:
    print(f"  bool({v!r:10}) = {bool(v)}")

print("\nEverything else is Truthy – e.g.:")
truthy_examples = [1, -1, 3.14, 'hi', [0], (None,), {'a': 1}]
for v in truthy_examples:
    print(f"  bool({v!r:10}) = {bool(v)}")
````

## 8. Lists

A **list** is an ordered, **mutable** sequence that can hold items of
**any type** (even mixed types).

| Property | Value |
|----------|-------|
| Ordered | Yes |
| Mutable | Yes |
| Allows duplicates | Yes |
| Syntax | `[item1, item2, ...]` |


````python
# Creating lists
empty   = []
numbers = [1, 2, 3, 4, 5]
mixed   = [42, "hello", 3.14, True, None]
nested  = [[1, 2], [3, 4], [5, 6]]
from_range = list(range(1, 11))

print(f"numbers   : {numbers}")
print(f"mixed     : {mixed}")
print(f"nested[1] : {nested[1]}")
print(f"from_range: {from_range}")
````

````python
# Indexing & slicing
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(f"First  : {fruits[0]}")
print(f"Last   : {fruits[-1]}")
print(f"Slice  : {fruits[1:4]}")
print(f"Reverse: {fruits[::-1]}")
````

````python
# Modifying lists
lst = [10, 20, 30]

lst.append(40)          # add at end
lst.insert(1, 15)       # insert at index
lst.extend([50, 60])    # extend with another iterable
print(f"After add  : {lst}")

lst.remove(15)          # remove first occurrence of value
popped = lst.pop()      # remove & return last item
del lst[0]              # delete by index
print(f"After remove: {lst}, popped={popped}")
````

````python
# List methods
nums = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"sorted   : {sorted(nums)}")
nums.sort()
print(f"sort()   : {nums}")
nums.reverse()
print(f"reverse(): {nums}")
print(f"count(1) : {nums.count(1)}")
print(f"index(9) : {nums.index(9)}")
print(f"min/max  : {min(nums)}, {max(nums)}")
print(f"sum      : {sum(nums)}")
````

````python
# List comprehensions
squares      = [x**2 for x in range(1, 11)]
evens        = [x for x in range(20) if x % 2 == 0]
flat_nested  = [n for row in [[1,2],[3,4],[5,6]] for n in row]

print(f"squares    : {squares}")
print(f"evens      : {evens}")
print(f"flattened  : {flat_nested}")
````

## 9. Tuples

A **tuple** is an ordered, **immutable** sequence.
Use tuples for data that should not change (coordinates, RGB, DB records).

| Property | Value |
|----------|-------|
| Ordered | Yes |
| Mutable | **No** |
| Allows duplicates | Yes |
| Syntax | `(item1, item2, ...)` |


````python
# Creating tuples
empty   = ()
single  = (42,)       # trailing comma required for single-item tuple!
point   = (3, 7)
rgb     = (255, 128, 0)
no_parens = 1, 2, 3  # parentheses are optional

print(f"single  : {single}  type={type(single).__name__}")
print(f"point   : {point}")
print(f"rgb     : {rgb}")
print(f"no_parens: {no_parens}")
````

````python
# Tuples are immutable – reading is fine, writing raises TypeError
coords = (10, 20, 30)
print(f"coords[0] = {coords[0]}")
print(f"coords[1:] = {coords[1:]}")

# Unpacking
x, y, z = coords
print(f"x={x}, y={y}, z={z}")

# coords[0] = 99   # TypeError: 'tuple' object does not support item assignment
print("Tuples are immutable (write blocked by comment).")
````

````python
# Named tuples – readable, self-documenting tuples
from collections import namedtuple

Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 7)
print(f"p     = {p}")
print(f"p.x   = {p.x}")
print(f"p.y   = {p.y}")
print(f"p[0]  = {p[0]}")   # also supports index access
````

## 10. Sets

A **set** is an **unordered** collection of **unique** elements.
Great for membership testing and eliminating duplicates.

| Property | Value |
|----------|-------|
| Ordered | No |
| Mutable | Yes (use `frozenset` for immutable) |
| Allows duplicates | **No** |
| Syntax | `{item1, item2, ...}` |


````python
# Creating sets
empty_set = set()          # NOT {} — that creates a dict!
nums   = {1, 2, 3, 4, 5}
dupes  = {1, 2, 2, 3, 3, 3}  # duplicates automatically removed
from_list = set([5, 4, 3, 2, 1])

print(f"nums     : {nums}")
print(f"dupes    : {dupes}  (duplicates removed!)")
print(f"from_list: {from_list}")
````

````python
# Set operations (mathematical)
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print(f"A | B  (union)       : {A | B}")
print(f"A & B  (intersection): {A & B}")
print(f"A - B  (difference)  : {A - B}")
print(f"A ^ B  (symmetric diff): {A ^ B}")
print(f"A <= B (subset?)     : {A <= B}")
````

````python
# Modifying sets
s = {10, 20, 30}
s.add(40)
s.update([50, 60, 70])
print(f"after add/update: {s}")

s.discard(99)    # no error if element missing
s.remove(10)     # raises KeyError if missing
print(f"after discard/remove: {s}")

# Fast membership test O(1)
print(f"20 in s : {20 in s}")
print(f"99 in s : {99 in s}")
````

## 11. Dictionaries

A **dictionary** (`dict`) is an ordered* mapping of **key-value pairs**.
Keys must be **unique** and **hashable** (str, int, tuple, etc.).

> *Insertion order preserved since Python 3.7+

| Property | Value |
|----------|-------|
| Ordered | Yes (Python 3.7+) |
| Mutable | Yes |
| Duplicate keys | No (last value wins) |
| Syntax | `{key: value, ...}` |


````python
# Creating dictionaries
empty  = {}
person = {
    "name"  : "Alice",
    "age"   : 30,
    "city"  : "Paris",
    "active": True,
}

# dict() constructor
config = dict(host="localhost", port=5432, db="mydb")

print(f"person : {person}")
print(f"config : {config}")
````

````python
# Accessing values
person = {"name": "Alice", "age": 30, "city": "Paris"}

print(person["name"])                   # direct access
print(person.get("age"))                # safer – returns None if missing
print(person.get("email", "N/A"))       # default value
````

````python
# Modifying dictionaries
d = {"a": 1, "b": 2}

d["c"] = 3          # add new key
d["a"] = 99         # update existing key
d.update({"d": 4, "e": 5})  # merge another dict
print(f"after update : {d}")

removed = d.pop("b")         # remove key, return value
print(f"popped b={removed}, d={d}")
````

````python
# Iterating over a dictionary
data = {"x": 10, "y": 20, "z": 30}

print("Keys   :", list(data.keys()))
print("Values :", list(data.values()))
print("Items  :", list(data.items()))

for key, value in data.items():
    print(f"  {key} => {value}")
````

````python
# Dictionary comprehensions
squares = {x: x**2 for x in range(1, 6)}
print(f"squares: {squares}")

words = ["apple", "banana", "cherry"]
word_len = {w: len(w) for w in words}
print(f"lengths: {word_len}")
````

## 12. None Type

`None` is Python's **null value** — it represents the absence of a value.
It is the only instance of the `NoneType` class.

Common uses:
- Default return value of functions that don't return anything
- Placeholder for optional variables
- Default argument values in function signatures


````python
# None basics
x = None
print(f"x = {x}")
print(f"type(x) = {type(x)}")

# Always use 'is' / 'is not' to compare with None (not ==)
print(f"x is None     : {x is None}")
print(f"x is not None : {x is not None}")
````

````python
# Functions without explicit return return None
def greet(name):
    print(f"Hello, {name}!")
    # no return statement

result = greet("Bob")
print(f"return value: {result!r}")  # None
````

````python
# None as default/optional parameter
def connect(host, port=None):
    if port is None:
        port = 5432   # set default only if not provided
    return f"Connecting to {host}:{port}"

print(connect("localhost"))
print(connect("db.example.com", 3306))
````

## 13. Type Checking

| Function | Use case |
|----------|---------|
| `type(obj)` | Returns exact type object |
| `isinstance(obj, type)` | Checks type (including subclasses) |


````python
# type() – returns the exact type
values = [42, 3.14, 'hi', True, [1, 2], (1,), {1}, {'a': 1}, None]
for v in values:
    print(f"  type({v!r:12}) = {type(v).__name__}")
````

````python
# isinstance() – checks type membership (preferred in production code)
x = 42
print(f"isinstance(42, int)        : {isinstance(x, int)}")
print(f"isinstance(42, float)      : {isinstance(x, float)}")
print(f"isinstance(42, (int,float)): {isinstance(x, (int, float))}")

# bool is a subclass of int
print(f"isinstance(True, int)      : {isinstance(True, int)}")
print(f"type(True) is int          : {type(True) is int}")  # False!
````

````python
# Practical type guard pattern
def double(value):
    if not isinstance(value, (int, float)):
        raise TypeError(f"Expected a number, got {type(value).__name__}")
    return value * 2

print(double(5))
print(double(3.14))
try:
    double("oops")
except TypeError as e:
    print(f"TypeError: {e}")
````

## 14. Type Conversion (Casting)

Python provides built-in functions to **explicitly convert** between types.

| Function | Converts to |
|----------|-------------|
| `int(x)` | Integer |
| `float(x)` | Float |
| `str(x)` | String |
| `bool(x)` | Boolean |
| `list(x)` | List |
| `tuple(x)` | Tuple |
| `set(x)` | Set |
| `dict(x)` | Dictionary |
| `complex(r, i)` | Complex |


````python
# int conversions
print(int(3.99))       # 3  (truncates, does NOT round)
print(int("42"))
print(int(True))       # 1
print(int("0xFF", 16)) # 255 (specify base)
print(int("0b1010", 2))# 10
````

````python
# float conversions
print(float(42))
print(float("3.14"))
print(float(True))     # 1.0
print(float("inf"))    # infinity
````

````python
# str conversions
print(str(42))
print(str(3.14))
print(str(True))
print(str([1, 2, 3]))
print(str(None))
````

````python
# Collection conversions
lst = [1, 2, 3, 2, 1]
print(f"list  -> tuple  : {tuple(lst)}")
print(f"list  -> set    : {set(lst)}  (no dupes!)")
print(f"str   -> list   : {list("hello")}")
print(f"range -> list   : {list(range(5))}")

# dict from list of pairs
pairs = [("a", 1), ("b", 2), ("c", 3)]
print(f"pairs -> dict   : {dict(pairs)}")
````

````python
# Implicit vs explicit conversion
# Python performs implicit conversion when safe
result = 5 + 2.0    # int + float -> float
print(f"5 + 2.0 = {result}  ({type(result).__name__})")

# But NOT between incompatible types
try:
    bad = "hello" + 5
except TypeError as e:
    print(f"TypeError: {e}")

# You must cast explicitly
good = "Value: " + str(5)
print(good)
````

## 1. Dictionary Basics & Creation

Keys must be hashable and immutable (strings, numbers, tuples). Values can be any Python object.


```python
# Creation methods
user = {"id": 101, "username": "ada_lovelace", "role": "admin"}
from_constructor = dict(name="Bob", age=30)
from_pairs = dict([("a", 1), ("b", 2)])

print(f"user:            {user}")
print(f"from_constructor:{from_constructor}")
print(f"from_pairs:      {from_pairs}")
```

## 2. Safe Access & In-Place Modification

Accessing a missing key with `dict[key]` raises `KeyError`. Use `.get(key, default)` or `.setdefault(key, default)` for safe access.


```python
config = {"host": "localhost", "port": 8000}

# 1. Safe access with fallback
timeout = config.get("timeout", 30)  # Returns default 30 without raising KeyError
print(f"Retrieved timeout: {timeout}")

# 2. setdefault: Returns value if key exists; sets and returns default if missing
log_level = config.setdefault("log_level", "INFO")
print(f"Config after setdefault: {config}")

# 3. Removing elements
port = config.pop("port")            # Removes and returns value
print(f"Popped port: {port}")
```

## 3. Dictionary Views & Iteration

Dictionary `.keys()`, `.values()`, and `.items()` return dynamic **view objects** that reflect changes to the dictionary in real-time.


```python
inventory = {"apples": 50, "bananas": 100, "cherries": 25}

print("Iterating over key-value pairs:")
for item, count in inventory.items():
    print(f"  -> {item:<10}: {count} in stock")

# Key check
print(f"Is 'apples' in inventory? {'apples' in inventory}")
```

## 4. Modern Dictionary Merge Operators (`|` and `|=`)

Introduced in Python 3.9 (PEP 584), the union operators allow clean dictionary merging.


```python
base_settings = {"theme": "dark", "font_size": 14, "auto_save": True}
user_overrides = {"font_size": 16, "show_line_numbers": True}

# Merge with | (right side values overwrite left side on collision)
merged = base_settings | user_overrides
print("Merged settings:", merged)
```

## 5. Dictionary Comprehensions

Transform and filter dictionaries with concise comprehension expressions.


```python
prices_usd = {"laptop": 1200, "mouse": 25, "monitor": 300, "cable": 10}

# 1. Apply 10% discount to all items >= $100
discounted = {k: v * 0.9 for k, v in prices_usd.items() if v >= 100}
print(f"Discounted items (>= $100): {discounted}")

# 2. Inverting a dictionary (swapping keys and values)
status_codes = {200: "OK", 404: "NOT_FOUND", 500: "SERVER_ERROR"}
inverted = {v: k for k, v in status_codes.items()}
print(f"Inverted status codes: {inverted}")
```

## 6. Specialized Dictionaries: `defaultdict` & `Counter`

The `collections` module provides high-performance specialized dictionary variants.


```python
from collections import defaultdict, Counter

# 1. defaultdict: automatically initializes missing keys with a default factory
grouped_words = defaultdict(list)
for word in ["apple", "apricot", "banana", "blueberry", "cherry"]:
    grouped_words[word[0]].append(word)

print(f"Grouped by initial: {dict(grouped_words)}")

# 2. Counter: multiset frequency counter
text = "to be or not to be that is the question"
word_counts = Counter(text.split())
print(f"Top 2 most common words: {word_counts.most_common(2)}")
```

## 7. Hands-On Interactive Challenges


```python
# Challenge 1: Deep merge two nested dictionaries
def deep_merge(dict1: dict, dict2: dict) -> dict:
    result = dict1.copy()
    for k, v in dict2.items():
        if k in result and isinstance(result[k], dict) and isinstance(v, dict):
            result[k] = deep_merge(result[k], v)
        else:
            result[k] = v
    return result

# Challenge 2: Group list of objects by key attribute
def group_by(records: list[dict], key: str) -> dict[str, list[dict]]:
    grouped = defaultdict(list)
    for record in records:
        grouped[record[key]].append(record)
    return dict(grouped)

# Automated verification tests
d1 = {"a": 1, "nested": {"x": 10, "y": 20}}
d2 = {"b": 2, "nested": {"y": 99, "z": 30}}
merged = deep_merge(d1, d2)

assert merged["a"] == 1
assert merged["b"] == 2
assert merged["nested"] == {"x": 10, "y": 99, "z": 30}

employees = [
    {"name": "Alice", "dept": "Eng"},
    {"name": "Bob", "dept": "Sales"},
    {"name": "Charlie", "dept": "Eng"}
]
grouped_emp = group_by(employees, "dept")
assert len(grouped_emp["Eng"]) == 2
assert len(grouped_emp["Sales"]) == 1

print("[OK] All Dictionary Challenges Passed!")
```

## 8. Quick Reference Card & Summary

| Method / Operator | Syntax | Purpose |
| :--- | :--- | :--- |
| **Get with Default** | `d.get(k, default)` | Safe retrieval without `KeyError` |
| **Set Default** | `d.setdefault(k, default)` | Returns existing or inserts default |
| **Items View** | `d.items()` | Yields `(key, value)` pairs |
| **Pop** | `d.pop(k)` | Removes key and returns value |
| **Merge Operator** | `d1 \| d2` | Returns new merged dictionary |
| **In-place Merge** | `d1 \|= d2` | Updates d1 with d2 in-place |

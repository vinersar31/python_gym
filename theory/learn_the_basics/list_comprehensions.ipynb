## 1. Foundations & Syntax Anatomy

### 📐 The Basic Formula
```python
[expression for item in iterable]
```

Let's compare a traditional `for` loop with a list comprehension:


````python
# 1. Traditional for loop accumulator
squares_loop = []
for x in range(1, 6):
    squares_loop.append(x ** 2)

# 2. List comprehension equivalent
squares_comp = [x ** 2 for x in range(1, 6)]

print(f"Loop result:          {squares_loop}")
print(f"Comprehension result: {squares_comp}")
print(f"Are results identical? {squares_loop == squares_comp}")
````

## 2. Conditional Filtering & If-Else Transformations

### 🔍 1. Filtering (`if` at the end)
`[expression for item in iterable if condition]` (Selects only items where condition is True).

### 🔀 2. Transformation (`if-else` before `for`)
`[expr_if_true if condition else expr_if_false for item in iterable]` (Transforms all items).


````python
numbers = [-5, 12, -3, 8, 0, 15, -1, 4]

# Filtering: Keep only positive even numbers
pos_evens = [x for x in numbers if x > 0 and x % 2 == 0]
print(f"Positive even numbers: {pos_evens}")

# Transformation: Replace negative numbers with 0, leave positive numbers intact
clamped = [x if x >= 0 else 0 for x in numbers]
print(f"Clamped non-negative:  {clamped}")

# Multi-condition categorization
labels = ["POSITIVE" if x > 0 else "NEGATIVE" if x < 0 else "ZERO" for x in numbers]
print(f"Value labels:          {labels}")
````

## 3. Multi-Loop & Nested Comprehensions

### 📦 1. Flattening 2D Matrices
`[item for row in matrix for item in row]` (Loops read left-to-right exactly like nested `for` loops).

### 🔄 2. Matrix Transposition
`[[row[i] for row in matrix] for i in range(cols)]`


````python
# Flattening a 2D matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

flattened = [val for row in matrix for val in row]
print(f"Flattened 2D matrix: {flattened}")

# Cartesian product of colors and sizes
colors = ["Red", "Blue"]
sizes = ["S", "M", "L"]
combinations = [(c, s) for c in colors for s in sizes]
print(f"Cartesian product (Total: {len(combinations)}): {combinations}")

# Matrix Transposition (swap rows and columns)
transposed = [[row[i] for row in matrix] for i in range(len(matrix[0]))]
print(f"Transposed matrix:   {transposed}")
````

## 4. Dictionary, Set, & Generator Comprehensions

Python extends comprehension syntax to dictionaries, sets, and generator expressions:

| Type | Syntax | Result |
| :--- | :--- | :--- |
| **List Comprehension** | `[x for x in it]` | `list` |
| **Set Comprehension** | `{x for x in it}` | `set` (Unique elements) |
| **Dict Comprehension**| `{k: v for x in it}`| `dict` (Key-value mapping) |
| **Generator Expression**| `(x for x in it)` | `generator` (Lazy $O(1)$ memory) |


````python
import sys

# 1. Set Comprehension (Automatic deduplication and normalization)
raw_tags = ["Python", "PYTHON", "fastapi", "FastAPI", "web", "WEB"]
unique_tags = {tag.lower() for tag in raw_tags}
print(f"Set comprehension:  {unique_tags}")

# 2. Dictionary Comprehension (Inverting key-value pairs)
id_to_user = {101: "Alice", 102: "Bob", 103: "Charlie"}
user_to_id = {name: uid for uid, name in id_to_user.items()}
print(f"Dict inversion:     {user_to_id}")

# 3. Generator Expression (Lazy Memory vs Immediate List Memory)
gen_exp = (x ** 2 for x in range(1_000_000))
list_comp = [x ** 2 for x in range(10_000)]

print(f"Generator memory:   {sys.getsizeof(gen_exp)} bytes (Streams 1M items in O(1) space!)")
print(f"List (10k) memory:  {sys.getsizeof(list_comp)} bytes")
````

## 5. Advanced Optimization with the Walrus Operator (`:=`)

The **walrus operator** (`:=`, Python 3.8+) allows assignment within an expression. This avoids calculating an expensive function twice in the filter and output expression.


````python
def expensive_computation(n: int) -> int:
    # Simulate mathematical calculation
    return (n ** 3) - (n ** 2)

# Without Walrus: expensive_computation(x) is called twice!
# result = [expensive_computation(x) for x in range(10) if expensive_computation(x) > 50]

# With Walrus: computed only once per item and reused!
optimized = [res for x in range(10) if (res := expensive_computation(x)) > 50]

print(f"Values > 50 (computed efficiently): {optimized}")
````

## 6. Performance Benchmarking & Bytecode Mechanics

### 🚀 Why are List Comprehensions Faster Than `for` Loops?
In a standard `for` loop, calling `.append()` requires a Python method lookup on every single iteration.
List comprehensions are executed in CPython using a specialized, optimized **`LIST_APPEND`** bytecode instruction that directly appends to the internal C array buffer without method lookup overhead.


````python
import timeit
import dis

# Speed comparison over 1 million iterations
setup = "nums = range(100)"

t_loop = timeit.timeit("res = []\nfor x in nums:\n    res.append(x * 2)", setup=setup, number=10_000)
t_comp = timeit.timeit("res = [x * 2 for x in nums]", setup=setup, number=10_000)
t_map  = timeit.timeit("res = list(map(lambda x: x * 2, nums))", setup=setup, number=10_000)

print(f"For Loop time:         {t_loop * 1000:.2f} ms")
print(f"List Comprehension:    {t_comp * 1000:.2f} ms ({t_loop / t_comp:.2f}x faster!)")
print(f"map() with lambda:     {t_map * 1000:.2f} ms")
````

## 7. Real-World Case Studies: Data Cleaning & Normalization

### 📊 Cleaning Dirty API / CSV Records


````python
raw_users = [
    {"id": "1", "name": " Alice ", "email": "ALICE@EXAMPLE.COM", "active": "true"},
    {"id": "2", "name": "Bob", "email": "invalid_email", "active": "false"},
    {"id": "3", "name": " Charlie", "email": "charlie@web.org", "active": "true"},
]

# Clean, normalize, and filter active verified users in one clean comprehension
clean_active_users = [
    {
        "id": int(u["id"]),
        "name": u["name"].strip(),
        "email": u["email"].strip().lower()
    }
    for u in raw_users
    if u["active"] == "true" and "@" in u["email"]
]

print("Cleaned Active Users:")
for user in clean_active_users:
    print(" ", user)
````

## 8. Common Pitfalls & Anti-Patterns

### ❌ Pitfall 1: The "One-Liner Monster"
*Anti-Pattern*: Writing deeply nested, 5-clause comprehensions that nobody can read.
*Fix*: If a comprehension spans more than 2 lines or has multiple nested conditions, use a standard `for` loop.

### ❌ Pitfall 2: Comprehensions for Pure Side Effects
*Anti-Pattern*: `[print(x) for x in items]` (allocates an unused list in memory).
*Fix*: Use a standard `for x in items: print(x)`.

### ❌ Pitfall 3: Forgetting Generator Expressions for Aggregations
*Anti-Pattern*: `sum([x**2 for x in range(10_000_000)])` (allocates hundreds of MBs of memory).
*Fix*: `sum(x**2 for x in range(10_000_000))` (zero extra memory).

## 9. Hands-On Interactive Challenges


````python
# Challenge 1: Transpose any NxM matrix using list comprehensions
def transpose_matrix(matrix: list[list[int]]) -> list[list[int]]:
    if not matrix or not matrix[0]:
        return []
    return [[row[col_idx] for row in matrix] for col_idx in range(len(matrix[0]))]

# Challenge 2: Invert dictionary and group keys sharing identical values
def invert_and_group(data: dict[str, int]) -> dict[int, list[str]]:
    unique_vals = {v for v in data.values()}
    return {val: [k for k, v in data.items() if v == val] for val in unique_vals}

# Challenge 3: Extract palindromic words of minimum length
def extract_palindromes(text: str, min_len: int = 3) -> list[str]:
    words = [w.strip(".,!?:;\"'").lower() for w in text.split()]
    return [w for w in words if len(w) >= min_len and w == w[::-1]]

# Automated verification tests
# Test 1: Matrix Transposition (3x2 -> 2x3)
mat = [[1, 2], [3, 4], [5, 6]]
assert transpose_matrix(mat) == [[1, 3, 5], [2, 4, 6]]

# Test 2: Invert and group
scores = {"Alice": 90, "Bob": 80, "Charlie": 90, "Diana": 80, "Evan": 100}
grouped_scores = invert_and_group(scores)
assert sorted(grouped_scores[90]) == ["Alice", "Charlie"]
assert sorted(grouped_scores[80]) == ["Bob", "Diana"]
assert grouped_scores[100] == ["Evan"]

# Test 3: Palindromes
sample_text = "Madam saw a racecar and said wow at noon on the radar"
pals = extract_palindromes(sample_text, min_len=4)
assert pals == ["madam", "racecar", "noon", "radar"]

print("[OK] All List Comprehension Challenges Passed!")
````

| Pattern | Syntax | Example |
| :--- | :--- | :--- |
| **Simple Mapping** | `[expr for x in it]` | `[x*2 for x in range(5)]` |
| **Filtering** | `[expr for x in it if cond]` | `[x for x in nums if x > 0]` |
| **If-Else Transform** | `[a if cond else b for x in it]` | `[x if x > 0 else 0 for x in nums]` |
| **Nested Flatten** | `[y for row in m for y in row]` | `[x for row in matrix for x in row]` |
| **Set Comprehension** | `{expr for x in it}` | `{x.lower() for x in tags}` |
| **Dict Comprehension**| `{k: v for x in it}` | `{u.id: u.name for u in users}` |
| **Generator Exp** | `(expr for x in it)` | `(x**2 for x in range(1_000_000))` |
| **Walrus Cache** | `[y for x in it if (y:=f(x)) > 0]`| `[res for x in it if (res:=calc(x)) > 0]` |

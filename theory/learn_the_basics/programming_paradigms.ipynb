## 1. Foundations: What is a Programming Paradigm?

A paradigm is a mental model for structuring software:
- **Imperative / Procedural**: *"Step 1 do this, Step 2 change that variable, Step 3 repeat."* (Focus on state mutation & subroutines).
- **Functional**: *"Transform data using pure mathematical functions without mutating state."* (Focus on data pipelines & immutability).
- **Object-Oriented**: *"Model the problem domain as interacting objects bundling state and behavior."* (Focus on encapsulation & polymorphism).
- **Declarative**: *"Describe what the desired result is, letting the language handle the how."* (Focus on comprehensions, queries, and pattern matching).

## 2. Procedural & Imperative Programming

In procedural programming, a program is structured as a sequence of procedural steps and subroutines that manipulate mutable data structures.


````python
# Procedural style: Sequential procedures modifying shared state
inventory = [
    {"name": "Laptop", "stock": 5, "price": 1000.0},
    {"name": "Mouse", "stock": 0, "price": 25.0},
    {"name": "Monitor", "stock": 12, "price": 300.0}
]

def restock_item(items: list[dict], item_name: str, quantity: int):
    for item in items:
        if item["name"] == item_name:
            item["stock"] += quantity
            return True
    return False

def calculate_inventory_value(items: list[dict]) -> float:
    total = 0.0
    for item in items:
        total += item["stock"] * item["price"]
    return total

# Execute procedures
restock_item(inventory, "Mouse", 10)
total_val = calculate_inventory_value(inventory)
print(f"[Procedural] Updated Mouse Stock: {inventory[1]['stock']}")
print(f"[Procedural] Total Inventory Value: ${total_val:,.2f}")
````

## 3. Functional Programming (FP) in Python

### 🔑 Core Principles of Functional Programming
1. **Pure Functions**: Identical inputs always produce identical outputs; zero side-effects.
2. **Immutability**: Data structures are never mutated in-place; transformations return new objects.
3. **First-Class & Higher-Order Functions**: Functions can be passed as arguments, returned, and composed.
4. **Tools**: `map()`, `filter()`, `functools.reduce()`, `functools.partial`, `operator` module.


````python
import functools
import operator
from typing import NamedTuple

# Immutable data structure
class Item(NamedTuple):
    name: str
    stock: int
    price: float

items = (
    Item("Laptop", 5, 1000.0),
    Item("Mouse", 0, 25.0),
    Item("Monitor", 12, 300.0)
)

# Pure functions
def is_in_stock(item: Item) -> bool:
    return item.stock > 0

def get_item_value(item: Item) -> float:
    return item.stock * item.price

# Pure Functional Pipeline (Filter -> Map -> Reduce)
in_stock_items = filter(is_in_stock, items)
item_values = map(get_item_value, in_stock_items)
total_val_fp = functools.reduce(operator.add, item_values, 0.0)

print(f"[Functional] In-Stock Total Inventory Value: ${total_val_fp:,.2f}")
````

## 4. Object-Oriented Programming (OOP) in Python

In OOP, state and behavior are encapsulated together within classes, modeling the domain through abstraction, inheritance, and polymorphism.


````python
from abc import ABC, abstractmethod

class Product(ABC):
    def __init__(self, name: str, stock: int, price: float):
        self.name = name
        self.stock = stock
        self.price = price

    @abstractmethod
    def calculate_tax(self) -> float:
        pass

    @property
    def total_value(self) -> float:
        return self.stock * self.price

class PhysicalProduct(Product):
    def calculate_tax(self) -> float:
        return self.total_value * 0.08  # 8% sales tax

class DigitalProduct(Product):
    def calculate_tax(self) -> float:
        return self.total_value * 0.02  # 2% digital service tax

catalog: list[Product] = [
    PhysicalProduct("Laptop", 5, 1000.0),
    DigitalProduct("E-Book", 100, 15.0)
]

for p in catalog:
    print(f"[OOP] {p.name:<10} | Value: ${p.total_value:>8,.2f} | Tax: ${p.calculate_tax():>6,.2f}")
````

## 5. Declarative & Pattern-Matching Styles

In declarative programming, we specify **what** result we want rather than orchestrating step-by-step control loops (e.g. list comprehensions and Python 3.10+ `match/case`).


````python
# Declarative transaction categorization with match/case
transactions = [
    {"type": "DEPOSIT", "amount": 500.0},
    {"type": "WITHDRAWAL", "amount": 120.0},
    {"type": "FEE", "amount": 5.0, "reason": "OVERDRAFT"},
    {"type": "TRANSFER", "amount": 300.0, "recipient": "Alice"}
]

def process_transaction(tx: dict) -> str:
    match tx:
        case {"type": "DEPOSIT", "amount": amt}:
            return f"Deposited ${amt:.2f}"
        case {"type": "WITHDRAWAL", "amount": amt} if amt > 100:
            return f"Large withdrawal: ${amt:.2f} (Alert Sent)"
        case {"type": "WITHDRAWAL", "amount": amt}:
            return f"Withdrew ${amt:.2f}"
        case {"type": "FEE", "amount": amt, "reason": r}:
            return f"Fee of ${amt:.2f} assessed for: {r}"
        case {"type": "TRANSFER", "amount": amt, "recipient": to}:
            return f"Transferred ${amt:.2f} to {to}"
        case _:
            return "Unknown transaction structure"

for tx in transactions:
    print(f"[Declarative] {process_transaction(tx)}")
````

## 6. Side-by-Side Comparison: Solving One Problem 4 Ways

**The Problem**: Given a list of order items, calculate the final total by filtering out cancelled items, applying a 10% discount to orders over $100, and adding 5% tax.

Let's solve it across all 4 paradigms:


````python
raw_orders = [
    {"item": "Desk", "price": 250.0, "status": "ACTIVE"},
    {"item": "Chair", "price": 80.0, "status": "CANCELLED"},
    {"item": "Lamp", "price": 40.0, "status": "ACTIVE"},
]

# ---------------------------------------------------------
# Approach 1: Procedural
# ---------------------------------------------------------
def calculate_procedural(orders: list[dict]) -> float:
    subtotal = 0.0
    for order in orders:
        if order["status"] == "ACTIVE":
            subtotal += order["price"]
    if subtotal > 100.0:
        subtotal *= 0.90  # 10% discount
    total = subtotal * 1.05  # 5% tax
    return round(total, 2)

# ---------------------------------------------------------
# Approach 2: Functional (Pure Functions + Expressions)
# ---------------------------------------------------------
def apply_discount(amount: float) -> float:
    return amount * 0.90 if amount > 100.0 else amount

def apply_tax(amount: float) -> float:
    return amount * 1.05

def calculate_functional(orders: list[dict]) -> float:
    active_prices = (o["price"] for o in orders if o["status"] == "ACTIVE")
    subtotal = sum(active_prices)
    return round(apply_tax(apply_discount(subtotal)), 2)

# ---------------------------------------------------------
# Approach 3: Modern Hybrid (The Pythonic Way)
# ---------------------------------------------------------
from dataclasses import dataclass

@dataclass(frozen=True)
class Order:
    item: str
    price: float
    status: str

    @property
    def is_active(self) -> bool:
        return self.status == "ACTIVE"

order_objs = [Order(**o) for o in raw_orders]
subtotal_hybrid = sum(o.price for o in order_objs if o.is_active)
final_hybrid = round((subtotal_hybrid * 0.90 if subtotal_hybrid > 100 else subtotal_hybrid) * 1.05, 2)

print(f"Procedural Result: ${calculate_procedural(raw_orders)}")
print(f"Functional Result: ${calculate_functional(raw_orders)}")
print(f"Hybrid Result:     ${final_hybrid}")
print(f"All Approaches Match: {calculate_procedural(raw_orders) == calculate_functional(raw_orders) == final_hybrid}")
````

## 7. Paradigm Trade-Offs & Decision Matrix

| Paradigm | Strengths | Weaknesses | Best Fit in Python |
| :--- | :--- | :--- | :--- |
| **Procedural** | Simple, intuitive, direct | Hard to maintain as codebase grows, shared mutable state | Small scripts, batch jobs, linear algorithms |
| **Functional** | Zero side-effects, highly testable, concurrency-safe | Can feel unnatural for stateful systems, recursion limits | Data pipelines, mathematical transformations, stream filtering |
| **Object-Oriented** | Domain encapsulation, code reusability, polymorphism | Over-engineering risk, tight coupling | Large domain models, GUI frameworks, web server backends |
| **Declarative** | Highly readable, concise, focuses on intent | Less control over fine-grained step execution | Comprehensions, querying, routing, configuration |

## 8. Common Pitfalls & Anti-Patterns

### ❌ Pitfall 1: Dogmatic OOP ("Java Syndrome")
*Anti-Pattern*: Creating a class just to hold a single method.
```python
# Bad: Unnecessary class boilerplate
class StringReverser:
    def reverse(self, s: str) -> str:
        return s[::-1]
```
*Fix*: Use a simple standalone function `def reverse_string(s: str) -> str: return s[::-1]`.

### ❌ Pitfall 2: "Fake Functional" Impure Functions
*Anti-Pattern*: Writing functions that take inputs but mutate external global state.
*Fix*: Pure functions must only compute and return new values without side effects.

## 9. Hands-On Interactive Challenges


````python
import functools
import operator

# Challenge 1: Pure Functional Pipeline
# Transform a list of numbers: keep only even positive numbers, square them, and sum them up
def functional_even_square_sum(numbers: list[int]) -> int:
    """Pure functional pipeline using filter, map, and reduce."""
    evens = filter(lambda x: x > 0 and x % 2 == 0, numbers)
    squares = map(lambda x: x ** 2, evens)
    return functools.reduce(operator.add, squares, 0)

# Challenge 2: Multi-Paradigm Strategy Pattern
# OOP Class with functional discount callable strategies
class PriceCalculator:
    def __init__(self, strategy_func):
        self.strategy_func = strategy_func

    def calculate(self, base_price: float) -> float:
        return round(self.strategy_func(base_price), 2)

# Functional strategy closures
def vip_discount(discount_pct: float):
    return lambda price: price * (1.0 - discount_pct)

# Automated verification tests
# Test 1: Functional pipeline (2^2 + 4^2 = 4 + 16 = 20)
assert functional_even_square_sum([-4, -2, 0, 1, 2, 3, 4, 5]) == 20
assert functional_even_square_sum([1, 3, 5]) == 0

# Test 2: Strategy pattern
calc_vip = PriceCalculator(strategy_func=vip_discount(0.20))
calc_standard = PriceCalculator(strategy_func=lambda p: p)

assert calc_vip.calculate(100.0) == 80.0
assert calc_standard.calculate(100.0) == 100.0

print("[OK] All Programming Paradigm Challenges Passed!")
````

## 10. Quick Reference Card & Summary Matrix

### 🎯 The Pythonic Paradigm Synthesis

> *"Python is pragmatic, not dogmatic."*

- Use **Functions** for stateless transformations and procedures.
- Use **Classes** (`@dataclass`, custom classes) when state and behavior are inherently tied together.
- Use **Comprehensions & Generators** for concise declarative data transformation.
- Use **Pure Functions** for deterministic business logic to ensure testability.

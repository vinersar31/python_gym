## 1. Basic Variable & Function Annotations

Annotate variables, parameters, and return types.


```python
# Variable annotations
age: int = 25
name: str = "Alice"
is_active: bool = True

def calculate_bmi(weight_kg: float, height_m: float) -> float:
    """Calculates Body Mass Index."""
    return round(weight_kg / (height_m ** 2), 2)

bmi = calculate_bmi(70.0, 1.75)
print(f"Calculated BMI: {bmi}")
```

## 2. Modern Built-in Generics (Python 3.9+)

Python 3.9+ allows using standard collection types (`list`, `dict`, `tuple`, `set`) directly as generics.


```python
# Built-in generic annotations
usernames: list[str] = ["alice", "bob", "charlie"]
user_scores: dict[str, int] = {"alice": 95, "bob": 88}
coordinate: tuple[float, float] = (37.7749, -122.4194)
unique_ids: set[int] = {101, 102, 103}

print(f"usernames:   {usernames}")
print(f"user_scores: {user_scores}")
```

## 3. Union Types (`int | str`) & Optional Values

Python 3.10+ uses the `|` syntax for Union types and Optional values (`str | None`).


```python
def find_user_id(username: str) -> int | None:
    database = {"alice": 1, "bob": 2}
    return database.get(username)

print(f"alice id:   {find_user_id('alice')}")
print(f"unknown id: {find_user_id('charlie')}")
```

## 4. Callable, Iterable, & Sequence Types

From the `typing` and `collections.abc` modules.


```python
from collections.abc import Callable, Iterable, Sequence

def apply_transformation(numbers: Iterable[int], transform_func: Callable[[int], int]) -> list[int]:
    return [transform_func(x) for x in numbers]

def square(n: int) -> int:
    return n * n

results = apply_transformation([1, 2, 3, 4], square)
print("Transformed results:", results)
```

## 5. Structured Dictionaries with `TypedDict`

Enforces strict key names and value types in dictionaries (e.g. JSON payloads).


```python
from typing import TypedDict

class UserProfile(TypedDict):
    id: int
    name: str
    email: str
    is_verified: bool

user_data: UserProfile = {
    "id": 1001,
    "name": "Grace Hopper",
    "email": "grace@navy.mil",
    "is_verified": True
}

print("TypedDict profile:", user_data)
```

## 6. Generic Functions & `TypeVar`

Writing generic functions that preserve argument and return type relationships.


```python
from typing import TypeVar, Sequence

T = TypeVar("T")

def get_first_element(items: Sequence[T]) -> T | None:
    return items[0] if items else None

first_num = get_first_element([10, 20, 30])      # Returns int
first_str = get_first_element(["a", "b", "c"])  # Returns str

print(f"first_num: {first_num} ({type(first_num).__name__})")
print(f"first_str: {first_str} ({type(first_str).__name__})")
```

## 7. Structural Subtyping with `typing.Protocol`


```python
from typing import Protocol

class Serializable(Protocol):
    def to_json(self) -> str: ...

class Product:
    def __init__(self, name: str, price: float):
        self.name = name
        self.price = price

    def to_json(self) -> str:
        return f'{{"name": "{self.name}", "price": {self.price}}}'

def export_data(item: Serializable) -> str:
    return item.to_json()

p = Product("Headphones", 99.99)
print("Exported JSON:", export_data(p))
```

## 8. Runtime Type Introspection with `get_type_hints()`


```python
import typing

class ServerConfig:
    host: str
    port: int
    debug_mode: bool

hints = typing.get_type_hints(ServerConfig)
print("Inspected type hints:", hints)
```

## 9. Hands-On Interactive Challenges


```python
from typing import TypeVar

T = TypeVar("T")

# Challenge 1: Generic binary search with type hints
def binary_search(sorted_seq: list[T], target: T) -> int | None:
    low, high = 0, len(sorted_seq) - 1
    while low <= high:
        mid = (low + high) // 2
        if sorted_seq[mid] == target:
            return mid
        elif sorted_seq[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return None

# Automated verification test
nums = [10, 20, 30, 40, 50]
assert binary_search(nums, 30) == 2
assert binary_search(nums, 99) is None

print("[OK] Type Annotation Challenge Passed!")
```

## 10. Quick Reference Card & Summary

| Type Pattern | Modern Syntax (3.10+) | Meaning |
| :--- | :--- | :--- |
| **Generic List** | `list[int]` | Homogeneous list of integers |
| **Generic Dict** | `dict[str, float]` | Mapping string keys to float values |
| **Union** | `int \| str` | Either int or str |
| **Optional** | `str \| None` | Either str or None |
| **Function Type** | `Callable[[int, int], bool]` | Function taking 2 ints, returning bool |
| **Protocol** | `class P(Protocol):` | Duck typing interface contract |

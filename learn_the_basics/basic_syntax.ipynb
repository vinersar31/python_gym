Python is **case-sensitive** everywhere — variable names, function names, class names, and keywords.
`message` and `Message` are **two distinct variables**.

## 2. Python Identifiers

An **identifier** is the name given to variables, functions, classes, modules, etc.

### ✅ Rules for valid identifiers

| Rule | Details |
|------|--------|
| Allowed characters | Letters (a-z, A-Z), digits (0-9), underscore (_) |
| Starting character | Must begin with a **letter** or **underscore** – never a digit |
| Forbidden | `@`, `$`, `%`, spaces, hyphens |
| Case-sensitive | `myVar` ≠ `myvar` ≠ `MYVAR` |

### 🎨 Naming conventions (PEP 8)

| Style | Used for | Example |
|-------|----------|---------|
| `snake_case` | variables, functions | `user_name`, `get_data()` |
| `CamelCase` | classes | `StudentRecord`, `HttpClient` |
| `ALL_CAPS` | constants | `MAX_SIZE`, `PI` |
| `_single_leading` | internal-use signal | `_helper` |
| `__double_leading` | name mangling in classes | `__private` |


```python
# ✅ Valid identifiers
age = 25
user_name = "Alice"
_private = "internal"
MAX_LIMIT = 100
myVar2 = 3.14
print(f"age={age}, user_name={user_name}, MAX_LIMIT={MAX_LIMIT}")
```

## 3. Reserved Keywords

Python's **reserved words** have special meaning and **cannot** be used as identifiers.


```python
import keyword
all_keywords = keyword.kwlist
print(f"Python has {len(all_keywords)} reserved keywords:\n")
cols = 5
for i, kw in enumerate(all_keywords):
    print(f"{kw:<12}", end="")
    if (i + 1) % cols == 0:
        print()
```

```python
# Check whether a word is a keyword
words_to_check = ["for", "hello", "class", "python", "True", "none"]
for word in words_to_check:
    status = "🔴 KEYWORD" if keyword.iskeyword(word) else "🟢 valid identifier"
    print(f"  '{word}' → {status}")
```

## 4. Lines and Indentation

Python uses **indentation** (not `{}`) to define code blocks.

- Standard: **4 spaces** per level (PEP 8)
- All lines in the same block must have **identical** indentation
- Wrong indentation → `IndentationError`


```python
score = 85

if score >= 90:
    print("Grade: A")       # 4-space indent
elif score >= 75:
    print("Grade: B")
    print("Good job!")      # same block as above
else:
    print("Grade: C or below")

print("Outside the if/elif/else block")  # back to base level
```

```python
# Nested indentation
for i in range(1, 4):          # outer loop – 1st level
    for j in range(1, 4):      # inner loop – 2nd level
        if i == j:             # condition  – 3rd level
            print(f"i={i}, j={j} → diagonal!")  # 4th level
        else:
            print(f"i={i}, j={j}")
```

```python
# ❌ IndentationError demo – uncomment to see the error
# x = 10
# if x > 5:
# print("No indent!")    # IndentationError: expected an indented block
print("IndentationError example is safely commented out.")
```

## 5. Multi-line Statements

| Method | Syntax | Best for |
|--------|--------|----------|
| Explicit continuation | Backslash `\` at end of line | Long expressions |
| Implicit continuation | Open bracket `(`, `[`, or `{` | Lists, dicts, calls |


```python
# Explicit continuation with backslash
total = 10 + 20 + \
        30 + 40 + \
        50
print(f"Total (backslash): {total}")
```

```python
# Implicit continuation inside brackets (preferred)
fruits = [
    "apple",
    "banana",
    "cherry",
    "date",
]

result = max(
    10, 25, 7, 42, 3
)

message = ("This is a very long string that "
           "spans multiple lines using "
           "implicit continuation.")

print(f"Fruits : {fruits}")
print(f"Max    : {result}")
print(f"Message: {message}")
```

## 6. Quotation in Python

| Quote type | Syntax | Typical use |
|------------|--------|-------------|
| Single quotes | `'...'` | Short strings |
| Double quotes | `"..."` | Short strings |
| Triple quotes | `'''...'''` or `"""..."""` | Multi-line, docstrings |


```python
word1 = 'hello'
word2 = "world"

# Useful when the string itself contains a quote character
sentence1 = "It's a beautiful day"       # apostrophe inside double quotes
sentence2 = 'He said "Python is great!"'  # double quotes inside single quotes

print(word1, word2)
print(sentence1)
print(sentence2)
```

```python
# Triple quotes – multi-line strings
poem = """Roses are red,
Violets are blue,
Python is awesome,
And so are you!"""
print(poem)
```

```python
# Triple quotes as docstrings
def greet(name):
    """Return a personalized greeting.

    Args:
        name (str): The name of the person to greet.
    Returns:
        str: A greeting string.
    """
    return f"Hello, {name}!"

print(greet("Alice"))
print("\nDocstring:")
print(greet.__doc__)
```

## 7. Comments

Comments are **ignored by the interpreter** – they explain code to humans.

| Type | Syntax | Notes |
|------|--------|-------|
| Single-line | `# comment` | From `#` to end of line |
| Inline | `code  # comment` | After a statement |
| Multi-line | Triple-quoted string | Technically a string literal |


```python
# This is a single-line comment
pi = 3.14159   # inline comment: approximation of π

radius = 5
# Multiply by 2 to get diameter before computing circumference
circumference = 2 * pi * radius
print(f"Circumference: {circumference:.4f}")
```

```python
# Multi-line block using consecutive single-line comments:
# This is line 1 of a multi-line comment.
# This is line 2.
# This is line 3.

# Triple-quoted string acting as a block comment (not assigned → discarded)
'''
This block of text is a string literal.
Since it is not assigned to a variable,
Python discards it – effectively acting as a comment.
'''

print("Comments have no effect on program output.")
```

```python
# Using comments to disable code during debugging
x = 10
y = 0
# result = x / y   # ZeroDivisionError – disable while debugging
result = x / (y + 1)   # safe fallback
print(f"Result: {result}")
```

## 8. Multiple Statements on One Line

Use a **semicolon** `;` to place multiple statements on a single line.

> ⚠️ PEP 8 discourages this – use it sparingly for readability.


```python
# Semicolons – legal but generally not recommended
a = 1; b = 2; c = 3
print(a, b, c)

# Preferred PEP 8 style:
x = 10
y = 20
z = 30
print(x, y, z)
```

```python
# Compound statements on one line (only for very simple cases)
number = 7
if number % 2 != 0: print(f"{number} is odd")

for i in range(5): print(i, end=' ')
print()
```

```python
# Ternary expression – Pythonic single-line conditional
age = 20
status = "adult" if age >= 18 else "minor"
print(f"Status: {status}")
```

## 9. Command Line Arguments

Pass arguments to a script from the terminal via `sys.argv`.

| `sys.argv` index | Value |
|-----------------|-------|
| `sys.argv[0]` | Script filename |
| `sys.argv[1]` | First argument |
| `sys.argv[2]` | Second argument |


```python
import sys
print(f"Script/kernel name : {sys.argv[0]}")
print(f"Number of arguments: {len(sys.argv)}")
print(f"All argv entries   : {sys.argv}")
```

```python
# Simulating: python script.py Alice 30
simulated_args = ["script.py", "Alice", "30"]

script_name = simulated_args[0]
name_arg    = simulated_args[1] if len(simulated_args) > 1 else "World"
age_arg     = simulated_args[2] if len(simulated_args) > 2 else "unknown"

print(f"Script : {script_name}")
print(f"Hello, {name_arg}! You are {age_arg} years old.")
```

```python
# Modern approach: argparse (standard library)
import argparse

parser = argparse.ArgumentParser(description="Example CLI script")
parser.add_argument("--name",  type=str, default="World", help="Your name")
parser.add_argument("--count", type=int, default=1,       help="Repetitions")

# parse_known_args is safe to call inside a Jupyter notebook
args, _ = parser.parse_known_args()

for _ in range(args.count):
    print(f"Hello, {args.name}!")
```

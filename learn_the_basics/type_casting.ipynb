## 1. Implicit vs. Explicit Type Conversion

- **Implicit Conversion**: Python automatically promotes smaller types (e.g. `int` + `float` $ightarrow$ `float`) without data loss.
- **Explicit Casting**: Developer explicitly calls constructor functions (`int()`, `float()`, `str()`, etc.).


```python
# Implicit type conversion
integer_val = 10
float_val = 3.5
result = integer_val + float_val  # Automatically converted to float!

print(f"result: {result} | type: {type(result).__name__}")

# Explicit type casting
str_num = "100"
converted_int = int(str_num)
print(f"converted_int: {converted_int} | type: {type(converted_int).__name__}")
```

## 2. Primitive Type Casting

Converting between `int`, `float`, and `str`.


```python
# Float to Int (Truncates toward zero)
print("int(3.99):  ", int(3.99))
print("int(-3.99): ", int(-3.99))

# String to Float / Int
print("float('3.1415'):", float("3.1415"))
print("int('42'):       ", int("42"))

# Numbers with different bases
print("int('1010', base=2):  ", int("1010", 2))   # Binary 10
print("int('FF', base=16):    ", int("FF", 16))    # Hex 255
```

## 3. Truthiness & Boolean Evaluation Rules

In Python, every object has an inherent boolean value.
- **Falsy Values**: `False`, `None`, `0`, `0.0`, `""`, `[]`, `()`, `{}`, `set()`.
- **Truthy Values**: Everything else!


```python
falsy_samples = [False, None, 0, 0.0, "", [], (), {}, set()]

print("Falsy values evaluation:")
for item in falsy_samples:
    print(f"  bool({repr(item):<8}) -> {bool(item)}")

# String 'False' is truthy!
print(f"\nbool('False'): {bool('False')} (Non-empty string is ALWAYS True!)")
```

## 4. Collection Type Conversions

Transforming data across `list`, `tuple`, `set`, and `dict`.


```python
# 1. Deduplicating a list via set
raw_items = [1, 2, 2, 3, 4, 4, 5]
unique_items = list(set(raw_items))
print("Deduplicated list:", unique_items)

# 2. Creating dict from paired tuples or zip()
keys = ["id", "name", "role"]
values = [101, "Alice", "Admin"]
user_dict = dict(zip(keys, values))
print("Zipped dictionary:", user_dict)
```

## 5. Base Conversions & Binary Encoding

Working with binary, hexadecimal, ASCII, and Unicode code points.


```python
number = 42

print(f"Binary  (bin): {bin(number)}")
print(f"Hex     (hex): {hex(number)}")
print(f"Octal   (oct): {oct(number)}")

# Character and Unicode conversions
char = 'A'
code_point = ord(char)
back_to_char = chr(code_point)
print(f"ord('A'): {code_point} -> chr(65): '{back_to_char}'")
```

## 6. Custom Object Type Conversion Protocols

Custom classes can define dunder conversion protocols (`__int__`, `__float__`, `__bool__`, `__str__`).


```python
class Fraction:
    def __init__(self, numerator: int, denominator: int):
        self.numerator = numerator
        self.denominator = denominator

    def __float__(self) -> float:
        return self.numerator / self.denominator

    def __int__(self) -> int:
        return int(float(self))

    def __bool__(self) -> bool:
        return self.numerator != 0

    def __str__(self) -> str:
        return f"{self.numerator}/{self.denominator}"

f = Fraction(7, 2)
print(f"Fraction: {f}")
print(f"float(f):  {float(f)}")
print(f"int(f):    {int(f)}")
print(f"bool(f):   {bool(f)}")
```

## 7. Common Pitfalls & Conversion Edge Cases

- `int("3.14")` raises `ValueError` (must cast to `float` first: `int(float("3.14"))`).
- `bool("False")` evaluates to `True` (non-empty string).

## 8. Hands-On Interactive Challenges


```python
# Challenge 1: Parse string to boolean safely ("true", "1", "yes" -> True)
def safe_parse_bool(val) -> bool:
    if isinstance(val, bool):
        return val
    if isinstance(val, (int, float)):
        return val != 0
    if isinstance(val, str):
        return val.strip().lower() in {"true", "1", "yes", "t", "y"}
    return False

# Challenge 2: Coerce heterogeneous numeric strings to float, with fallback
def safe_to_float(val, fallback: float = 0.0) -> float:
    try:
        return float(val)
    except (ValueError, TypeError):
        return fallback

# Automated verification tests
assert safe_parse_bool("True") is True
assert safe_parse_bool("yes") is True
assert safe_parse_bool("False") is False
assert safe_parse_bool(0) is False

assert safe_to_float("3.14") == 3.14
assert safe_to_float("invalid", fallback=99.0) == 99.0

print("[OK] All Type Casting Challenges Passed!")
```

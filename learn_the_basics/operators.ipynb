## 1. Arithmetic Operators

Used to perform **mathematical operations** on numeric values.

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `5 + 3` | `8` |
| `-` | Subtraction | `5 - 3` | `2` |
| `*` | Multiplication | `5 * 3` | `15` |
| `/` | True Division | `7 / 2` | `3.5` |
| `//` | Floor Division | `7 // 2` | `3` |
| `%` | Modulo | `7 % 2` | `1` |
| `**` | Exponentiation | `2 ** 8` | `256` |
| `-x` | Negation (unary) | `-5` | `-5` |


````python
a, b = 17, 5

print(f"{a} + {b}   = {a + b}")
print(f"{a} - {b}   = {a - b}")
print(f"{a} * {b}   = {a * b}")
print(f"{a} / {b}   = {a / b}")    # always returns float
print(f"{a} // {b}  = {a // b}")   # floor: rounds toward -inf
print(f"{a} % {b}   = {a % b}")    # remainder
print(f"{a} ** {b}  = {a ** b}")   # power
print(f"-{a}       = {-a}")         # unary negation
````

````python
# Floor division vs truncation — they differ for negatives!
print(f"-7 // 2  = {-7 // 2}")   # -4  (floor: rounds toward -inf)
print(f"int(-7/2)= {int(-7/2)}") # -3  (truncate: rounds toward 0)

# Modulo follows floor division
print(f"-7 % 2   = {-7 % 2}")    # 1   (not -1)

# divmod() returns both quotient and remainder at once
q, r = divmod(17, 5)
print(f"divmod(17, 5) = ({q}, {r})")  # (3, 2)
````

````python
# Arithmetic on other types

# String repetition & concatenation
print("ha" * 3)           # hahaha
print("Hello" + " " + "World")

# List repetition & concatenation
print([0] * 5)            # [0, 0, 0, 0, 0]
print([1, 2] + [3, 4])   # [1, 2, 3, 4]

# abs(), round(), pow()
print(f"abs(-42)       = {abs(-42)}")
print(f"round(3.14159, 2) = {round(3.14159, 2)}")
print(f"pow(2, 10)     = {pow(2, 10)}")
print(f"pow(2, 10, 100)= {pow(2, 10, 100)}")  # modular exponentiation
````

## 2. Comparison Operators

Compare two values and return a **Boolean** (`True` / `False`).

| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal to | `5 == 5` → `True` |
| `!=` | Not equal to | `5 != 3` → `True` |
| `>` | Greater than | `7 > 3` → `True` |
| `<` | Less than | `3 < 7` → `True` |
| `>=` | Greater than or equal | `5 >= 5` → `True` |
| `<=` | Less than or equal | `4 <= 5` → `True` |


````python
x, y = 10, 20

print(f"x == y  : {x == y}")
print(f"x != y  : {x != y}")
print(f"x >  y  : {x >  y}")
print(f"x <  y  : {x <  y}")
print(f"x >= y  : {x >= y}")
print(f"x <= y  : {x <= y}")
````

````python
# Chained comparisons (Pythonic!)
score = 75
print(f"60 <= score < 80  : {60 <= score < 80}")   # True
print(f"0 < 1 < 2 < 3    : {0 < 1 < 2 < 3}")      # True
print(f"1 < 2 > 1 < 3    : {1 < 2 > 1 < 3}")      # True

# Comparing strings (lexicographic order)
print(f"'apple' < 'banana': {'apple' < 'banana'}")
print(f"'Z' < 'a'        : {'Z' < 'a'}")  # uppercase < lowercase
````

````python
# == vs is  (value vs identity)
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(f"a == b  : {a == b}")   # True  (same value)
print(f"a is b  : {a is b}")   # False (different objects)
print(f"a is c  : {a is c}")   # True  (same object)

# Small int caching gotcha
p = 256
q = 256
print(f"256: p is q : {p is q}")  # True  (CPython caches -5..256)
p = 1000
q = 1000
print(f"1000: p is q: {p is q}")  # False (outside cache range)
````

## 3. Assignment Operators

Assign values to variables. The compound forms combine assignment with
an arithmetic or bitwise operation.

| Operator | Equivalent to | Example |
|----------|---------------|---------|
| `=` | — | `x = 5` |
| `+=` | `x = x + n` | `x += 3` |
| `-=` | `x = x - n` | `x -= 3` |
| `*=` | `x = x * n` | `x *= 3` |
| `/=` | `x = x / n` | `x /= 3` |
| `//=` | `x = x // n` | `x //= 3` |
| `%=` | `x = x % n` | `x %= 3` |
| `**=` | `x = x ** n` | `x **= 3` |
| `&=` | `x = x & n` | `x &= 3` |
| `|=` | `x = x | n` | `x |= 3` |
| `^=` | `x = x ^ n` | `x ^= 3` |
| `>>=` | `x = x >> n` | `x >>= 1` |
| `<<=` | `x = x << n` | `x <<= 1` |


````python
x = 16
print(f"initial   x = {x}")
x += 4;  print(f'x += 4    -> {x}')
x -= 2;  print(f'x -= 2    -> {x}')
x *= 3;  print(f'x *= 3    -> {x}')
x /= 6;  print(f'x /= 6    -> {x}')
x = int(x)
x //= 2; print(f'x //= 2   -> {x}')
x **= 3; print(f'x **= 3   -> {x}')
x %= 10; print(f'x %= 10   -> {x}')
````

````python
# Multiple assignment & unpacking (recap from Variables notebook)
a = b = c = 0
print(f"a={a}, b={b}, c={c}")

x, y, z = 1, 2, 3
print(f"x={x}, y={y}, z={z}")

# Swap without temp variable
x, y = y, x
print(f"After swap: x={x}, y={y}")
````

## 4. Logical Operators

Combine Boolean expressions. Python uses **words** (not symbols) for logical ops.

| Operator | Description | Returns |
|----------|-------------|--------|
| `and` | True if **both** operands are truthy | The first falsy operand, or the last one |
| `or` | True if **at least one** operand is truthy | The first truthy operand, or the last one |
| `not` | Inverts truthiness | `True` or `False` |

### Truth table

| A | B | A and B | A or B | not A |
|---|---|---------|--------|-------|
| T | T | T | T | F |
| T | F | F | T | F |
| F | T | F | T | T |
| F | F | F | F | T |


````python
# Basic logical operations
a, b = True, False
print(f"True  and True  = {True  and True}")
print(f"True  and False = {True  and False}")
print(f"False and True  = {False and True}")
print(f"False or  True  = {False or  True}")
print(f"False or  False = {False or  False}")
print(f"not True        = {not True}")
print(f"not False       = {not False}")
````

````python
# Short-circuit evaluation
# 'and' stops at the FIRST falsy value
# 'or'  stops at the FIRST truthy value

print(0 and 1/0)      # 0  (never evaluates 1/0)
print(1 or  1/0)      # 1  (never evaluates 1/0)

# 'and'/'or' return operands, not just True/False
print(0     and "hello")   # 0      (first falsy)
print("hi"  and "world")   # world  (last, since all truthy)
print(None  or  "default") # default (first truthy)
print("val" or  "default") # val    (already truthy)
````

````python
# Practical short-circuit patterns

# Default value pattern
user_input = ""
name = user_input or "Anonymous"
print(f"name = {name}")

# Guard pattern (safe access)
data = None
length = data and len(data)   # avoids TypeError
print(f"length = {length}")

# Conditional expression with logical operators
x = 5
print(f"x>0 and x<10 : {x > 0 and x < 10}")
print(f"x<0 or  x>3  : {x < 0 or  x > 3}")
````

````python
# not operator
lst = []
print(f"not lst  (empty list): {not lst}")
lst = [1, 2, 3]
print(f"not lst  (full list) : {not lst}")

# Prefer 'is not None' over 'not x is None'
val = 0      # 0 is falsy, but it IS a valid value!
print(f"not val          : {not val}")
print(f"val is not None  : {val is not None}")
````

## 5. Bitwise Operators

Operate on integers **bit by bit** at the binary level.

| Operator | Name | Description |
|----------|------|-------------|
| `&` | AND | 1 if **both** bits are 1 |
| `|` | OR | 1 if **either** bit is 1 |
| `^` | XOR | 1 if bits are **different** |
| `~` | NOT | Inverts all bits (`~x = -x - 1`) |
| `<<` | Left shift | Shifts bits left (multiply by 2^n) |
| `>>` | Right shift | Shifts bits right (divide by 2^n) |


````python
a = 0b1100   # 12
b = 0b1010   # 10

print(f"a      = {a:4d}  ({a:08b})")
print(f"b      = {b:4d}  ({b:08b})")
print(f"a & b  = {a & b:4d}  ({(a & b):08b})  AND")
print(f"a | b  = {a | b:4d}  ({(a | b):08b})  OR")
print(f"a ^ b  = {a ^ b:4d}  ({(a ^ b):08b})  XOR")
print(f"~a     = {~a:4d}  ({~a & 0xFF:08b})  NOT")
print(f"a << 2 = {a << 2:4d}  ({(a << 2):08b})  Left shift")
print(f"a >> 1 = {a >> 1:4d}  ({(a >> 1):08b})  Right shift")
````

````python
# Practical uses of bitwise operators

# Check if a number is even (fast)
for n in range(8):
    parity = "even" if (n & 1) == 0 else "odd"
    print(f"  {n} is {parity}")
````

````python
# Bit shifting as fast multiply/divide by powers of 2
x = 8
print(f"{x} << 1 = {x << 1}  (x * 2)")
print(f"{x} << 2 = {x << 2}  (x * 4)")
print(f"{x} >> 1 = {x >> 1}  (x / 2)")
print(f"{x} >> 2 = {x >> 2}  (x / 4)")

# Flag/permission system using bit masks
READ    = 0b001   # 1
WRITE   = 0b010   # 2
EXECUTE = 0b100   # 4

permissions = READ | WRITE      # grant read + write
print(f"\nPermissions: {permissions:03b}")
print(f"Can read?   : {bool(permissions & READ)}")
print(f"Can write?  : {bool(permissions & WRITE)}")
print(f"Can execute?: {bool(permissions & EXECUTE)}")
````

## 6. Membership Operators

Test whether a value is present in a **sequence** or **collection**.

| Operator | Description | Example |
|----------|-------------|--------|
| `in` | Returns `True` if value **is** in the sequence | `'a' in 'cat'` → `True` |
| `not in` | Returns `True` if value **is not** in the sequence | `5 not in [1,2,3]` → `True` |

Works on: `str`, `list`, `tuple`, `set`, `dict` (tests keys), `range`.


````python
# Strings
print(f"'Py' in 'Python'    : {'Py' in 'Python'}")
print(f"'xyz' in 'Python'  : {'xyz' in 'Python'}")
print(f"'z' not in 'Python': {'z' not in 'Python'}")
````

````python
# Lists & Tuples
nums = [1, 2, 3, 4, 5]
print(f"3 in nums     : {3 in nums}")
print(f"9 in nums     : {9 in nums}")
print(f"9 not in nums : {9 not in nums}")

coords = (10, 20, 30)
print(f"20 in coords  : {20 in coords}")
````

````python
# Sets (O(1) average — much faster than list for large data)
valid_users = {"alice", "bob", "carol"}
username = "bob"
print(f"{username!r} in valid_users: {username in valid_users}")

# Dicts (tests KEYS by default)
config = {"host": "localhost", "port": 5432}
print(f""host" in config   : {"host" in config}")
print(f""timeout" in config: {"timeout" in config}")
print(f"5432 in config.values(): {5432 in config.values()}")
````

````python
# Membership in range
r = range(0, 100, 5)   # 0, 5, 10, ..., 95
print(f"25 in range(0,100,5) : {25 in r}")
print(f"27 in range(0,100,5) : {27 in r}")

# Practical: filter a list
allowed = {"admin", "editor", "viewer"}
roles   = ["admin", "guest", "editor", "hacker", "viewer"]
valid = [r for r in roles if r in allowed]
print(f"Valid roles: {valid}")
````

## 7. Identity Operators

Test whether two variables refer to the **same object** in memory
(not just equal values).

| Operator | Description |
|----------|--------------|
| `is` | `True` if both point to **the same object** (`id(a) == id(b)`) |
| `is not` | `True` if both point to **different objects** |

> **Rule of thumb:** Use `==` for value equality, `is` only for `None`,
> `True`, and `False`.


````python
# is vs ==
a = [1, 2, 3]
b = [1, 2, 3]   # same content, different object
c = a           # same object

print(f"a == b  : {a == b}")    # True  (equal values)
print(f"a is b  : {a is b}")    # False (different objects)
print(f"a is c  : {a is c}")    # True  (same object)
print(f"id(a)   : {id(a)}")
print(f"id(b)   : {id(b)}")
print(f"id(c)   : {id(c)}")
````

````python
# The correct way: use 'is' for singletons
x = None
print(f"x is None     : {x is None}")       # correct
print(f"x == None     : {x == None}")       # works but frowned upon
print(f"x is not None : {x is not None}")

# True and False are singletons
print(f"True  is True : {True  is True}")
print(f"False is False: {False is False}")
````

````python
# CPython small integer cache (-5 to 256)
a = 100; b = 100
print(f"100: a is b = {a is b}")   # True  (cached)

a = 300; b = 300
print(f"300: a is b = {a is b}")   # False (not cached)

# String interning
s1 = "hello"
s2 = "hello"
s3 = "hel" + "lo"   # may or may not be interned
print(f"s1 is s2 : {s1 is s2}")   # usually True (interned literals)
print(f"s1 == s2 : {s1 == s2}")   # always True

print("\nLesson: rely on == for value comparison, is only for None/singletons.")
````

## 8. Operator Precedence

When an expression contains multiple operators, Python evaluates them in
**precedence order** (highest to lowest):

| Priority | Operator(s) | Description |
|----------|-------------|-------------|
| 1 (highest) | `()` | Parentheses |
| 2 | `**` | Exponentiation (right-to-left!) |
| 3 | `+x`, `-x`, `~x` | Unary plus, minus, NOT |
| 4 | `*`, `/`, `//`, `%` | Multiplication, division, modulo |
| 5 | `+`, `-` | Addition, subtraction |
| 6 | `<<`, `>>` | Bit shifts |
| 7 | `&` | Bitwise AND |
| 8 | `^` | Bitwise XOR |
| 9 | `|` | Bitwise OR |
| 10 | `==`, `!=`, `>`, `<`, `>=`, `<=`, `is`, `in`, ... | Comparisons |
| 11 | `not` | Logical NOT |
| 12 | `and` | Logical AND |
| 13 (lowest) | `or` | Logical OR |


````python
# Precedence examples
print(f"2 + 3 * 4       = {2 + 3 * 4}")      # 14  (* before +)
print(f"(2 + 3) * 4     = {(2 + 3) * 4}")    # 20  (parens first)
print(f"2 ** 3 ** 2     = {2 ** 3 ** 2}")    # 512 (**  is right-assoc: 2**(3**2)=2**9)
print(f"(2 ** 3) ** 2   = {(2 ** 3) ** 2}")  # 64  
print(f"-2 ** 2         = {-2 ** 2}")         # -4  (** before unary -)
print(f"(-2) ** 2       = {(-2) ** 2}")       # 4
````

````python
# Logical operator precedence: not > and > or
print(f"True or  False and False = {True or  False and False}") # True  (and first)
print(f"not False and  True      = {not False and  True}")       # True  (not first)
print(f"not (False and True)     = {not (False and True)}")      # True

# Comparison chaining and precedence
x = 5
print(f"1 < x < 10         : {1 < x < 10}")
print(f"x > 0 and x < 10   : {x > 0 and x < 10}")  # equivalent
````

````python
# When in doubt: USE PARENTHESES
# Bad (relies on memorising precedence):
result_bad  = 2 + 3 * 4 ** 2 // 6 - 1

# Good (explicit):
result_good = 2 + ((3 * (4 ** 2)) // 6) - 1

print(f"bad  = {result_bad}")
print(f"good = {result_good}")
print(f"same? {result_bad == result_good}")
````

## 9. Walrus Operator (Python 3.8+)

The **walrus operator** `:=` (named for its resemblance to a walrus) is the
**assignment expression** — it assigns a value AND returns it in the same step.

```python
# Syntax
variable := expression
```

Useful in `while` loops, `if` statements, and comprehensions to avoid
computing the same value twice.


````python
# Traditional approach (compute twice)
import re
text = "Order number: 12345"

match = re.search(r'\d+', text)
if match:
    print(f"Found: {match.group()}")

# Walrus approach (compute once)
if m := re.search(r'\d+', text):
    print(f"Found (walrus): {m.group()}")
````

````python
# Walrus in a while loop
import random
random.seed(42)

# Traditional: read-test-process
rolls = []
while True:
    n = random.randint(1, 6)
    if n == 6:
        break
    rolls.append(n)
print(f"Traditional: {rolls}")

# Walrus: assign-and-test in one go
random.seed(42)
rolls_w = []
while (n := random.randint(1, 6)) != 6:
    rolls_w.append(n)
print(f"Walrus     : {rolls_w}")
````

````python
# Walrus in list comprehension (filter + transform without double call)
def expensive(x):
    """Simulate a slow computation."""
    return x ** 2

data = range(10)

# Without walrus: expensive() called twice
results_old = [expensive(x) for x in data if expensive(x) > 25]

# With walrus: expensive() called once per element
results_new = [y for x in data if (y := expensive(x)) > 25]

print(f"old: {results_old}")
print(f"new: {results_new}")
````

## 10. Quick Reference Card

All operator categories demonstrated in one place:


````python
# ==================================================================
# PYTHON OPERATORS – QUICK REFERENCE
# ==================================================================

a, b = 12, 5

# --- 1. Arithmetic ---
print('-- Arithmetic --')
print(a + b, a - b, a * b, a / b, a // b, a % b, a ** b)

# --- 2. Comparison ---
print('-- Comparison --')
print(a == b, a != b, a > b, a < b, a >= b, a <= b)

# --- 3. Augmented Assignment ---
print('-- Augmented Assignment --')
x = 10; x += 5; print(x)   # 15

# --- 4. Logical ---
print('-- Logical --')
print(True and False, True or False, not True)

# --- 5. Bitwise ---
print('-- Bitwise --')
print(a & b, a | b, a ^ b, ~a, a << 1, a >> 1)

# --- 6. Membership ---
print('-- Membership --')
print(5 in [1, 5, 9], 7 not in [1, 5, 9])

# --- 7. Identity ---
print('-- Identity --')
x = None; print(x is None, x is not None)

# --- 8. Walrus ---
print('-- Walrus --')
nums = [1, 4, 9, 16, 25]
big  = [y for n in nums if (y := n * 2) > 20]
print(big)
````

## Summary

| Category | Operators | Returns |
|----------|-----------|---------|
| **Arithmetic** | `+` `-` `*` `/` `//` `%` `**` | Number |
| **Comparison** | `==` `!=` `>` `<` `>=` `<=` | `bool` |
| **Assignment** | `=` `+=` `-=` `*=` `/=` `//=` `%=` `**=` `&=` `|=` `^=` `>>=` `<<=` | Variable updated |
| **Logical** | `and` `or` `not` | Operand or `bool` |
| **Bitwise** | `&` `|` `^` `~` `<<` `>>` | Integer |
| **Membership** | `in` `not in` | `bool` |
| **Identity** | `is` `is not` | `bool` |
| **Walrus** | `:=` | Assigned value |

### Key rules to remember
- `**` is **right-associative**: `2**3**2` = `2**(3**2)` = `512`
- `-2**2` = `-(2**2)` = `-4` — use `(-2)**2` for `4`
- `and`/`or` return **operands**, not just `True`/`False`
- Use `==` for value equality, `is` only for `None`/singletons
- When unsure about precedence → **add parentheses**

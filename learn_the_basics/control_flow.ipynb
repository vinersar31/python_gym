## 1. if / elif / else

The `if` statement evaluates a condition and executes a block only when the
condition is **truthy**.

```python
if condition:
    # runs when condition is True
elif other_condition:
    # runs when other_condition is True
else:
    # runs when none of the above matched
```

- `elif` and `else` are **optional**.
- You can have **any number** of `elif` branches.
- Only the **first** matching branch executes; the rest are skipped.


````python
# Basic if / elif / else
score = 78

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score {score} -> Grade {grade}")
````

````python
# if with no else – does nothing when condition is False
temperature = 35

if temperature > 30:
    print("It is hot outside!")

print("Program continues...")
````

````python
# Truthy / falsy conditions – any expression can be used
items = ["apple", "banana"]

if items:                          # non-empty list is truthy
    print(f"Cart has {len(items)} item(s)")
else:
    print("Cart is empty")

username = ""
if not username:                   # empty string is falsy
    print("Username cannot be empty")
````

````python
# Multiple conditions with logical operators
age    = 25
income = 55000

if age >= 18 and income >= 30000:
    print("Loan approved")
elif age >= 18 or income >= 50000:
    print("Conditional approval")
else:
    print("Loan denied")
````

## 2. Nested Conditionals

You can place `if` statements inside other `if` blocks.
Keep nesting **shallow** (max 2-3 levels) — use early returns or
`and`/`or` to flatten deep nesting.


````python
# Nested if
num = 42

if num > 0:
    if num % 2 == 0:
        print(f"{num} is positive and even")
    else:
        print(f"{num} is positive and odd")
elif num == 0:
    print("Number is zero")
else:
    print(f"{num} is negative")
````

````python
# Anti-pattern: deeply nested (hard to read)
def classify_deep(x):
    if x > 0:
        if x < 100:
            if x % 2 == 0:
                return "small positive even"
            else:
                return "small positive odd"
        else:
            return "large positive"
    else:
        return "non-positive"

# Better: flatten with early returns and combined conditions
def classify_flat(x):
    if x <= 0:         return "non-positive"
    if x >= 100:       return "large positive"
    if x % 2 == 0:     return "small positive even"
    return "small positive odd"

for v in [-5, 0, 7, 42, 150]:
    print(f"  {v:>4} -> deep: {classify_deep(v):<22} flat: {classify_flat(v)}")
````

## 3. Ternary Expression

A **one-line conditional** that returns a value. Use when the logic is simple.

```python
value_if_true  if  condition  else  value_if_false
```


````python
# Basic ternary
age = 20
status = "adult" if age >= 18 else "minor"
print(status)
````

````python
# Ternary in various contexts
x = -5

# Absolute value
abs_x = x if x >= 0 else -x
print(f"abs({x}) = {abs_x}")

# In a function return
def sign(n):
    return "positive" if n > 0 else ("negative" if n < 0 else "zero")

for n in [-3, 0, 7]:
    print(f"  sign({n:>2}) = {sign(n)}")
````

````python
# Ternary in f-strings and list comprehensions
numbers = range(-4, 5)

labels = [f"{n} (even)" if n % 2 == 0 else f"{n} (odd)" for n in numbers]
for label in labels:
    print(" ", label)
````

````python
# When NOT to use ternary: complex logic -> use if/else instead
score = 85

# Bad: hard to read
# grade = "A" if score>=90 else "B" if score>=80 else "C" if score>=70 else "D" if score>=60 else "F"

# Good: clear if/elif chain
if   score >= 90: grade = "A"
elif score >= 80: grade = "B"
elif score >= 70: grade = "C"
elif score >= 60: grade = "D"
else:             grade = "F"

print(f"Grade: {grade}")
````

## 4. match / case (Python 3.10+)

Python 3.10 introduced **structural pattern matching** — a powerful
alternative to long `if/elif` chains, supporting literal, sequence,
mapping, and class patterns.

```python
match subject:
    case pattern1:
        ...
    case pattern2:
        ...
    case _:       # wildcard – always matches (like 'default')
        ...
```


````python
import sys
print(f"Python version: {sys.version}")

# Only run match/case examples on Python 3.10+
if sys.version_info >= (3, 10):
    print("match/case is supported!")
else:
    print("match/case requires Python 3.10+ – skipping demo")
````

````python
# ---- Literal pattern matching ----
def http_status(code):
    match code:
        case 200: return "OK"
        case 201: return "Created"
        case 400: return "Bad Request"
        case 401: return "Unauthorized"
        case 403: return "Forbidden"
        case 404: return "Not Found"
        case 500: return "Internal Server Error"
        case _:   return "Unknown Status"

for code in [200, 201, 404, 500, 999]:
    print(f"  {code} -> {http_status(code)}")
````

````python
# ---- OR patterns  |  and guards ----
def day_type(day):
    match day.lower():
        case "saturday" | "sunday":
            return "Weekend"
        case "monday" | "tuesday" | "wednesday" | "thursday" | "friday":
            return "Weekday"
        case _:
            return "Unknown"

for d in ["Monday", "Saturday", "Sunday", "Holiday"]:
    print(f"  {d:<10} -> {day_type(d)}")
````

````python
# ---- Sequence patterns ----
def describe_point(point):
    match point:
        case (0, 0):
            return "Origin"
        case (0, y):
            return f"On Y-axis at y={y}"
        case (x, 0):
            return f"On X-axis at x={x}"
        case (x, y) if x == y:
            return f"On diagonal at ({x}, {y})"
        case (x, y):
            return f"Point at ({x}, {y})"

for pt in [(0,0), (0,5), (3,0), (4,4), (3,7)]:
    print(f"  {str(pt):<10} -> {describe_point(pt)}")
````

````python
# ---- Mapping (dict) patterns ----
def handle_command(command):
    match command:
        case {"action": "quit"}:
            return "Quitting..."
        case {"action": "move", "direction": direction}:
            return f"Moving {direction}"
        case {"action": "attack", "weapon": weapon}:
            return f"Attacking with {weapon}"
        case _:
            return "Unknown command"

commands = [
    {"action": "quit"},
    {"action": "move", "direction": "north"},
    {"action": "attack", "weapon": "sword"},
    {"action": "dance"},
]
for cmd in commands:
    print(f"  {str(cmd):<40} -> {handle_command(cmd)}")
````

## 5. for Loops

The `for` loop iterates over any **iterable** (list, tuple, str, range,
dict, set, file, generator, …).

```python
for variable in iterable:
    # body executed for each element
```


````python
# Iterating over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(f"  I like {fruit}")
````

````python
# range() – the most common for-loop tool
# range(stop)
print("range(5):", list(range(5)))

# range(start, stop)
print("range(2, 8):", list(range(2, 8)))

# range(start, stop, step)
print("range(0, 20, 3):", list(range(0, 20, 3)))

# Counting down
print("range(10, 0, -2):", list(range(10, 0, -2)))
````

````python
# enumerate() – get index AND value
colours = ["red", "green", "blue"]

for i, colour in enumerate(colours):
    print(f"  [{i}] {colour}")

print()
# Start index at 1
for i, colour in enumerate(colours, start=1):
    print(f"  {i}. {colour}")
````

````python
# zip() – iterate over multiple iterables in parallel
names  = ["Alice", "Bob", "Carol"]
scores = [92, 85, 78]
grades = ["A",  "B",  "C" ]

for name, score, grade in zip(names, scores, grades):
    print(f"  {name:<6} | {score} | {grade}")

print()
# zip stops at shortest – use zip_longest to keep going
from itertools import zip_longest
a = [1, 2, 3]
b = ["x", "y"]
for pair in zip_longest(a, b, fillvalue="?"):
    print(pair)
````

````python
# Iterating over dicts
person = {"name": "Alice", "age": 30, "city": "Paris"}

print("Keys:")
for key in person:              # iterates keys by default
    print(f"  {key}")

print("Values:")
for value in person.values():
    print(f"  {value}")

print("Items:")
for key, value in person.items():
    print(f"  {key}: {value}")
````

````python
# Iterating over a string (character by character)
for ch in "Python":
    print(ch, end=" ")
print()

# Nested for loops
print("Multiplication table (3x3):")
for i in range(1, 4):
    for j in range(1, 4):
        print(f"  {i}x{j}={i*j}", end="  ")
    print()
````

````python
# Useful built-ins with for loops
numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5]

print(f"min    : {min(numbers)}")
print(f"max    : {max(numbers)}")
print(f"sum    : {sum(numbers)}")
print(f"sorted : {sorted(numbers)}")
print(f"any >8 : {any(n > 8 for n in numbers)}")
print(f"all >0 : {all(n > 0 for n in numbers)}")

# reversed() and sorted()
print(f"reversed: {list(reversed(numbers))}")
````

## 6. while Loops

A `while` loop repeats a block **as long as** a condition is truthy.
Use it when the number of iterations is **not known in advance**.

```python
while condition:
    # body
```

> Always ensure the condition eventually becomes `False`, or use `break`
> to avoid an **infinite loop**.


````python
# Basic while loop
count = 0
while count < 5:
    print(f"  count = {count}")
    count += 1
print("Done")
````

````python
# Countdown
n = 5
while n > 0:
    print(f"  {n}...")
    n -= 1
print("  Lift off!")
````

````python
# while with user-input simulation (using a list as mock input)
mock_inputs = ["hello", "", "world", "quit"]
idx = 0

while True:
    # Simulate input()
    user = mock_inputs[idx]; idx += 1
    print(f"  >> {user!r}")

    if user == "quit":
        print("  Goodbye!")
        break
    if not user:
        print("  (empty input, skipping)")
        continue
    print(f"  You typed: {user.upper()}")
````

````python
# Classic algorithms with while

# Collatz conjecture
def collatz_steps(n):
    steps = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps += 1
    return steps

for start in [6, 11, 27]:
    print(f"  collatz({start:>2}) takes {collatz_steps(start)} steps")

# Binary search
def binary_search(arr, target):
    lo, hi = 0, len(arr) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if arr[mid] == target:   return mid
        elif arr[mid] < target:  lo = mid + 1
        else:                    hi = mid - 1
    return -1

data = list(range(0, 100, 5))
print(f"\n  binary_search({data}, 45) -> index {binary_search(data, 45)}")
print(f"  binary_search({data}, 43) -> index {binary_search(data, 43)}")
````

## 7. break, continue, pass

| Statement | Effect |
|-----------|--------|
| `break` | **Exit** the nearest enclosing loop immediately |
| `continue` | **Skip** the rest of the current iteration, jump to next |
| `pass` | **Do nothing** — syntactic placeholder |


````python
# --- break ---
# Exit loop when first even number > 5 found
numbers = [1, 3, 5, 7, 8, 11, 12]
for n in numbers:
    if n > 5 and n % 2 == 0:
        print(f"  Found it: {n}")
        break
    print(f"  Checking {n}...")
````

````python
# --- continue ---
# Print only odd numbers, skip even ones
for n in range(1, 11):
    if n % 2 == 0:
        continue     # skip the rest of this iteration
    print(f"  {n}", end=" ")
print()
````

````python
# --- pass ---
# Placeholder while you plan the implementation
def todo_function():
    pass   # will implement later

class EmptyClass:
    pass

# Silently ignore a specific error
try:
    result = 10 / 0
except ZeroDivisionError:
    pass   # intentionally ignored

# if branch as placeholder
flag = True
if flag:
    pass   # TODO: handle the True case
else:
    print("Flag is False")

print("pass examples ran with no errors")
````

````python
# break and continue in nested loops
# break only exits the NEAREST loop
print("Finding first pair that sums to 10:")
found = False
for i in range(1, 6):
    for j in range(1, 6):
        if i + j == 10:
            print(f"  Found: {i} + {j} = 10")
            found = True
            break         # exits inner loop only
    if found:
        break             # exits outer loop

print()
# Pythonic: use a function and return to break all loops at once
def find_pair(limit, target):
    for i in range(1, limit + 1):
        for j in range(1, limit + 1):
            if i + j == target:
                return i, j
    return None

print(f"  Pythonic result: {find_pair(5, 10)}")
````

## 8. Loop else Clause

Both `for` and `while` loops support an optional `else` clause.
The `else` block runs **only when the loop completes normally**
(i.e. was NOT terminated by `break`).

```python
for item in iterable:
    if condition:
        break
else:
    # runs only if break was never hit
    ...
```

This is very useful for **search loops** — run something if the target
was never found.


````python
# for...else: prime checker
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            break          # found a divisor -> not prime
    else:
        return True        # loop completed without break -> prime
    return False

primes = [n for n in range(2, 30) if is_prime(n)]
print(f"Primes < 30: {primes}")
````

````python
# for...else: search in a list
def find_admin(users):
    for user in users:
        if user["role"] == "admin":
            print(f"  Admin found: {user['name']}")
            break
    else:
        print("  No admin found in the list")

users_with_admin = [
    {"name": "Alice", "role": "user"},
    {"name": "Bob",   "role": "admin"},
    {"name": "Carol", "role": "user"},
]
users_no_admin = [
    {"name": "Dave",  "role": "user"},
    {"name": "Eve",   "role": "user"},
]

find_admin(users_with_admin)
find_admin(users_no_admin)
````

````python
# while...else
import random
random.seed(0)

attempts = 0
max_attempts = 5
target = 3

while attempts < max_attempts:
    roll = random.randint(1, 6)
    attempts += 1
    print(f"  Attempt {attempts}: rolled {roll}")
    if roll == target:
        print(f"  Got {target}!")
        break
else:
    print(f"  Did not roll {target} in {max_attempts} attempts")
````

## 9. Comprehensions

Comprehensions are concise, readable **one-line loops** that build collections.

| Type | Syntax | Returns |
|------|--------|---------|
| List | `[expr for x in it if cond]` | `list` |
| Dict | `{k: v for x in it if cond}` | `dict` |
| Set  | `{expr for x in it if cond}` | `set` |
| Generator | `(expr for x in it if cond)` | lazy `generator` |


````python
# --- List comprehensions ---
squares  = [x**2 for x in range(1, 11)]
evens    = [x for x in range(20) if x % 2 == 0]
words    = ["hello", "world", "python"]
upper    = [w.upper() for w in words]

print(f"squares : {squares}")
print(f"evens   : {evens}")
print(f"upper   : {upper}")
````

````python
# Nested list comprehension (flatten a matrix)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

flat      = [n for row in matrix for n in row]
transposed = [[row[i] for row in matrix] for i in range(3)]

print(f"flat      : {flat}")
print(f"transposed: {transposed}")
````

````python
# --- Dict comprehensions ---
# Square mapping
sq_map = {x: x**2 for x in range(1, 6)}
print(f"sq_map : {sq_map}")

# Invert a dict
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in original.items()}
print(f"inverted: {inverted}")

# Filter dict entries
scores = {"Alice": 92, "Bob": 58, "Carol": 76, "Dave": 45}
passed = {name: s for name, s in scores.items() if s >= 60}
print(f"passed  : {passed}")
````

````python
# --- Set comprehensions ---
sentence = "the quick brown fox jumps over the lazy dog"

# Unique lengths of words
unique_lengths = {len(word) for word in sentence.split()}
print(f"unique word lengths: {sorted(unique_lengths)}")

# Unique first letters
first_letters = {word[0] for word in sentence.split()}
print(f"unique first chars : {sorted(first_letters)}")
````

````python
# --- Generator expressions ---
# Like a list comprehension but lazy — values computed on demand

# Memory efficient for large data
gen = (x**2 for x in range(1, 1_000_001))
print(f"type    : {type(gen)}")
print(f"next()  : {next(gen)}")
print(f"next()  : {next(gen)}")

# Can be consumed once only
total = sum(x**2 for x in range(1, 101))   # no extra [] needed
print(f"sum of squares 1..100: {total}")

# any()/all() with generators (short-circuits!)
nums = [2, 4, 6, 8, 9, 10]
print(f"all even: {all(n % 2 == 0 for n in nums)}")
print(f"any > 5 : {any(n > 5 for n in nums)}")
````

## 10. Quick Reference Card

All control flow patterns in one place:


````python
# ==================================================================
# PYTHON CONTROL FLOW – QUICK REFERENCE
# ==================================================================

# --- if / elif / else ---
x = 42
if x > 100:    label = "big"
elif x > 10:   label = "medium"
else:          label = "small"
print(f"if/elif/else : {label}")

# --- ternary ---
parity = "even" if x % 2 == 0 else "odd"
print(f"ternary      : {x} is {parity}")

# --- for loop ---
total = 0
for n in range(1, 6):
    total += n
print(f"for loop     : sum 1..5 = {total}")

# --- while loop ---
n = 1
product = 1
while n <= 5:
    product *= n
    n += 1
print(f"while loop   : 5! = {product}")

# --- break / continue ---
odds_under_10 = []
for i in range(20):
    if i >= 10: break
    if i % 2 == 0: continue
    odds_under_10.append(i)
print(f"break/cont   : {odds_under_10}")

# --- loop else ---
for i in range(2, 10):
    if 15 % i == 0:
        print(f"for...else   : 15 is divisible by {i}")
        break
else:
    print("for...else   : no divisor found")

# --- comprehensions ---
squares  = [n**2  for n in range(1, 6)]
sq_map   = {n: n**2 for n in range(1, 6)}
sq_set   = {n**2  for n in range(1, 6)}
sq_gen   = sum(n**2 for n in range(1, 6))
print(f"list comp    : {squares}")
print(f"dict comp    : {sq_map}")
print(f"set comp     : {sq_set}")
print(f"gen expr sum : {sq_gen}")
````

## Summary

| Statement | Purpose | Key detail |
|-----------|---------|------------|
| `if/elif/else` | Branch on condition | Only first matching branch runs |
| `match/case` | Structural pattern matching | Python 3.10+, supports guards |
| Ternary `a if c else b` | Inline conditional value | Keep it simple |
| `for x in iterable` | Fixed iteration | Works on any iterable |
| `while condition` | Repeat until False | Guard against infinite loops |
| `break` | Exit loop early | Exits the **nearest** loop only |
| `continue` | Skip current iteration | Jumps to next loop step |
| `pass` | Do nothing | Syntactic placeholder |
| `for/while...else` | Post-loop action | Skipped if `break` was hit |
| `[x for x in it]` | List comprehension | Concise, readable |
| `{k:v for x in it}` | Dict comprehension | Key-value mapping |
| `{x for x in it}` | Set comprehension | Unique values |
| `(x for x in it)` | Generator expression | Lazy — memory efficient |

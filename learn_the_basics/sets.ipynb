## 1. Set Creation & Uniqueness

Sets are created with curly braces `{}` or the `set()` constructor. An empty set must be created with `set()`, as `{}` creates an empty dictionary.


```python
# Creation
numbers_set = {1, 2, 3, 4, 5, 2, 3}  # Duplicates automatically removed!
empty_set = set()                    # NOT {} which creates a dict!
from_string = set("MISSISSIPPI")

print(f"numbers_set (unique): {numbers_set}")
print(f"type(empty_set):      {type(empty_set)}")
print(f"from_string:          {from_string}")
```

## 2. Mathematical Set Operations & Operators

Sets provide both operator syntax and method syntax for mathematical set algebra.


```python
frontend_devs = {"Alice", "Bob", "Charlie", "Diana"}
backend_devs = {"Charlie", "Diana", "Evan", "Frank"}

# 1. Union (All unique developers): A | B
all_devs = frontend_devs | backend_devs
print(f"Union (|):                {all_devs}")

# 2. Intersection (Full-stack devs in both): A & B
full_stack = frontend_devs & backend_devs
print(f"Intersection (&):         {full_stack}")

# 3. Difference (Frontend ONLY): A - B
frontend_only = frontend_devs - backend_devs
print(f"Difference (-):           {frontend_only}")

# 4. Symmetric Difference (Specialized, not in both): A ^ B
specialized = frontend_devs ^ backend_devs
print(f"Symmetric Difference (^): {specialized}")

# 5. Subset & Superset tests
leads = {"Charlie", "Diana"}
print(f"Is leads subset of frontend? {leads.issubset(frontend_devs)}")
print(f"Is frontend superset of leads? {frontend_devs >= leads}")
```

## 3. Set Mutation Methods

Methods for adding and removing elements from sets.


```python
tags = {"python", "coding"}

tags.add("tutorial")             # Add single element: O(1)
tags.update(["web", "fastapi"])  # Add multiple elements

tags.discard("non_existent")     # Safe removal (no error if missing)
tags.remove("coding")            # Removes item (raises KeyError if missing)

print("Updated tags:", tags)
```

## 4. Set Comprehensions

Similar to list comprehensions, but wrapped in `{}` to produce unique sets.


```python
words = ["apple", "Banana", "APPLE", "cherry", "banana", "DATE"]

# Unique lowercase word lengths
unique_lengths = {len(w) for w in words}
normalized_words = {w.lower() for w in words}

print(f"Normalized words: {normalized_words}")
print(f"Unique lengths:   {unique_lengths}")
```

## 5. Immutable Sets with `frozenset`

Standard sets are mutable and unhashable. A `frozenset` is immutable and hashable, allowing it to be used as a dictionary key or an element inside another set.


```python
permissions_role_map = {
    frozenset(["read", "write"]): "Editor",
    frozenset(["read"]): "Viewer",
    frozenset(["read", "write", "delete", "admin"]): "SuperAdmin"
}

user_perms = frozenset(["read", "write"])
print(f"Assigned role for {user_perms}: {permissions_role_map[user_perms]}")
```

## 6. Performance Benchmark: Set vs. List Membership

Set membership testing `x in my_set` is $O(1)$ constant time vs $O(n)$ linear scan in lists.


```python
import timeit

size = 100_000
test_val = 99_999

large_list = list(range(size))
large_set = set(large_list)

t_list = timeit.timeit(f"{test_val} in large_list", globals=globals(), number=1_000)
t_set = timeit.timeit(f"{test_val} in large_set", globals=globals(), number=1_000)

print(f"List search time (1K lookups): {t_list*1000:.2f} ms")
print(f"Set search time  (1K lookups): {t_set*1000:.4f} ms")
print(f"Set lookup speedup: {t_list / t_set:.2f}x faster!")
```

## 7. Hands-On Interactive Challenges


```python
# Challenge 1: Find common elements across N lists
def find_common_elements(*lists: list) -> set:
    if not lists:
        return set()
    return set(lists[0]).intersection(*[set(l) for l in lists[1:]])

# Challenge 2: Detect if two words are anagrams using set counting
def have_same_character_set(word1: str, word2: str) -> bool:
    return set(word1.lower()) == set(word2.lower())

# Automated verification tests
assert find_common_elements([1, 2, 3, 4], [2, 3, 5], [3, 2, 9]) == {2, 3}
assert have_same_character_set("listen", "silent") is True
assert have_same_character_set("hello", "world") is False

print("[OK] All Set Challenges Passed!")
```

## 8. Quick Reference Card & Summary

| Operation | Operator | Method | Purpose |
| :--- | :--- | :--- | :--- |
| **Union** | `A \| B` | `A.union(B)` | Elements in either A or B |
| **Intersection** | `A & B` | `A.intersection(B)` | Elements in both A and B |
| **Difference** | `A - B` | `A.difference(B)` | Elements in A but not B |
| **Sym. Difference** | `A ^ B` | `A.symmetric_difference(B)` | Elements in A or B, but not both |
| **Subset** | `A <= B` | `A.issubset(B)` | Is every element of A in B? |
| **Superset** | `A >= B` | `A.issuperset(B)` | Does A contain every element of B? |

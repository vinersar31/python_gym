## 1. Tuple Creation & Single-Element Syntax

Tuples are created with parentheses `()` or comma separation.


```python
# Creation methods
empty_tuple = ()
point = (10, 20)
implicit_tuple = 1, 2, 3  # Parentheses are optional!

# Single-element tuple MUST have a trailing comma:
not_a_tuple = (42)    # int, not a tuple!
single_tuple = (42,)  # tuple of length 1

print(f"type((42)):   {type(not_a_tuple)}")
print(f"type((42,)):  {type(single_tuple)}")
print(f"point:        {point} (len: {len(point)})")
```

## 2. Tuple Packing & Extended Unpacking

Python supports elegant tuple packing and starred `*` extended unpacking (PEP 3132).


```python
# 1. Standard unpacking
x, y, z = (100, 200, 300)
print(f"x={x}, y={y}, z={z}")

# 2. Variable swap using tuple packing/unpacking
a, b = 5, 10
a, b = b, a  # Swapped without temporary variable!
print(f"Swapped: a={a}, b={b}")

# 3. Extended unpacking with *rest
first, *middle, last = [10, 20, 30, 40, 50, 60]
print(f"first:  {first}")
print(f"middle: {middle}")
print(f"last:   {last}")
```

## 3. Immutability Semantics & Mutable Contents

The tuple structure itself cannot be modified (cannot append, remove, or reassign indices). However, if a tuple contains a **mutable element** (like a list), that nested element can still be modified!


```python
nested_tuple = ("Alice", [90, 85, 95])

print("Initial tuple:", nested_tuple)

# Mutating the list INSIDE the tuple:
nested_tuple[1].append(100)
print("After appending to nested list:", nested_tuple)

# Attempting to reassign index raises TypeError:
try:
    nested_tuple[0] = "Bob"
except TypeError as e:
    print(f"[Immutability Protected] {e}")
```

## 4. Tuple Performance & Memory Optimization

CPython optimizes tuples with fixed memory structs and free-list caching, making them faster to allocate and more compact than lists.


```python
import sys
import timeit

t_data = (1, 2, 3, 4, 5)
l_data = [1, 2, 3, 4, 5]

print(f"Tuple size: {sys.getsizeof(t_data)} bytes")
print(f"List size:  {sys.getsizeof(l_data)} bytes")

# Instantiation speed benchmark
t_time = timeit.timeit("(1, 2, 3, 4, 5)", number=1_000_000)
l_time = timeit.timeit("[1, 2, 3, 4, 5]", number=1_000_000)

print(f"Tuple creation time (1M): {t_time*1000:.2f} ms")
print(f"List creation time  (1M): {l_time*1000:.2f} ms")
print(f"Speedup: {l_time / t_time:.2f}x faster with tuples")
```

## 5. Tuples as Dictionary Keys & Hashability

Because tuples containing immutable elements are **hashable**, they can be used as multi-dimensional coordinate keys in dictionaries.


```python
# Grid game coordinate board
board_grid = {
    (0, 0): "Start",
    (0, 1): "Treasure",
    (1, 0): "Monster",
    (1, 1): "Exit"
}

pos = (0, 1)
print(f"Event at coordinate {pos}: {board_grid[pos]}")
```

## 6. NamedTuples (`collections.namedtuple` & `typing.NamedTuple`)

NamedTuples combine the memory efficiency and immutability of tuples with named attribute access (`obj.name`).


```python
from typing import NamedTuple

# Modern typed NamedTuple
class GeoLocation(NamedTuple):
    city: str
    latitude: float
    longitude: float
    elevation_m: float = 0.0  # Default value

loc = GeoLocation("Tokyo", 35.6762, 139.6503, elevation_m=40.0)

print(f"City: {loc.city} | Lat: {loc.latitude} | Lon: {loc.longitude}")
print(f"Access by index (loc[0]): {loc[0]}")
print(f"As dictionary: {loc._asdict()}")
```

## 7. Hands-On Interactive Challenges


```python
# Challenge 1: Find min, max, and average returning a named tuple
class StatsResult(NamedTuple):
    minimum: float
    maximum: float
    average: float

def compute_stats(numbers: list[float]) -> StatsResult:
    if not numbers:
        raise ValueError("List cannot be empty.")
    return StatsResult(
        minimum=min(numbers),
        maximum=max(numbers),
        average=sum(numbers) / len(numbers)
    )

# Challenge 2: Partition a sequence into head, body, and tail
def partition_sequence(seq: tuple) -> tuple:
    first, *body, last = seq
    return first, tuple(body), last

# Automated verification tests
stats = compute_stats([10.0, 20.0, 30.0, 40.0, 50.0])
assert stats.minimum == 10.0
assert stats.maximum == 50.0
assert stats.average == 30.0

head, body, tail = partition_sequence((1, 2, 3, 4, 5))
assert head == 1
assert body == (2, 3, 4)
assert tail == 5

print("[OK] All Tuple Challenges Passed!")
```

## 8. Quick Reference Card & Summary

| Feature | Tuple | List |
| :--- | :--- | :--- |
| **Mutability** | Immutable (Fixed) | Mutable (Dynamic) |
| **Syntax** | `(1, 2, 3)` or `1, 2` | `[1, 2, 3]` |
| **Memory** | Compact fixed struct | Over-allocated dynamic buffer |
| **Hashability** | Hashable (can be dict key if elements are hashable) | Unhashable |
| **Primary Use** | Heterogeneous records, fixed data, return values | Homogeneous dynamic collections |

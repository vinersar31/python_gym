# Linked Lists in Python — Implementations & Algorithmic Patterns

A **Linked List** is a linear data structure where elements (nodes) are stored in non-contiguous memory locations.
Each node contains a **data value** and a **pointer (reference)** to the next (and optionally previous) node.

## 1. Linked Lists vs. Arrays

| Feature | Array | Linked List |
|---------|-------|-------------|
| **Memory Allocation** | Contiguous memory block | Non-contiguous nodes allocated on heap |
| **Index Lookup `arr[i]`** | $O(1)$ constant time | $O(n)$ linear traversal |
| **Prepend / Insert at Head** | $O(n)$ requires shifting elements | $O(1)$ constant pointer update |
| **Deletion at Head** | $O(n)$ requires shifting elements | $O(1)$ constant pointer update |
| **Deletion in Middle (given Node)** | $O(n)$ shifting | $O(1)$ pointer redirection |
| **Cache Locality** | Excellent (CPU cache prefetching) | Poor (pointer chasing across heap) |

## 2. Singly Linked List Implementation

A **Singly Linked List** node contains `data` and `next` pointer.


```python
class Node:
    def __init__(self, val):
        self.val = val
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None
        self.size = 0

    def prepend(self, val):  # O(1)
        new_node = Node(val)
        new_node.next = self.head
        self.head = new_node
        self.size += 1

    def append(self, val):   # O(n) without tail pointer, O(1) with tail
        new_node = Node(val)
        if not self.head:
            self.head = new_node
        else:
            curr = self.head
            while curr.next:
                curr = curr.next
            curr.next = new_node
        self.size += 1

    def delete(self, val):   # O(n)
        if not self.head: return False
        if self.head.val == val:
            self.head = self.head.next
            self.size -= 1
            return True
        curr = self.head
        while curr.next and curr.next.val != val:
            curr = curr.next
        if curr.next:
            curr.next = curr.next.next
            self.size -= 1
            return True
        return False

    def __iter__(self):
        curr = self.head
        while curr:
            yield curr.val
            curr = curr.next

    def __str__(self):
        return " -> ".join(str(val) for val in self) + " -> None"

# Testing SinglyLinkedList
sll = SinglyLinkedList()
sll.append(10)
sll.append(20)
sll.append(30)
sll.prepend(5)
print("Initial List:", sll)
sll.delete(20)
print("After deleting 20:", sll)
```

## 3. Doubly Linked List Implementation

A **Doubly Linked List** node contains `data`, `next`, and `prev` pointers,
allowing bidirectional traversal.


```python
class DoublyNode:
    def __init__(self, val):
        self.val = val
        self.prev = None
        self.next = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def append(self, val):
        new_node = DoublyNode(val)
        if not self.head:
            self.head = self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node

    def forward_str(self):
        res = []
        curr = self.head
        while curr:
            res.append(str(curr.val))
            curr = curr.next
        return " <-> ".join(res) + " <-> None"

    def backward_str(self):
        res = []
        curr = self.tail
        while curr:
            res.append(str(curr.val))
            curr = curr.prev
        return " <-> ".join(res) + " <-> None"

dll = DoublyLinkedList()
for x in [1, 2, 3, 4]: dll.append(x)
print("Forward :", dll.forward_str())
print("Backward:", dll.backward_str())
```

## 4. Circular Linked List Implementation

In a **Circular Linked List**, the `next` pointer of the last node points back to the `head` node.


```python
class CircularLinkedList:
    def __init__(self):
        self.head = None

    def append(self, val):
        new_node = Node(val)
        if not self.head:
            self.head = new_node
            new_node.next = self.head
        else:
            curr = self.head
            while curr.next != self.head:
                curr = curr.next
            curr.next = new_node
            new_node.next = self.head

    def display(self, limit=7):
        if not self.head: return "Empty"
        res = []
        curr = self.head
        count = 0
        while count < limit:
            res.append(str(curr.val))
            curr = curr.next
            count += 1
        return " -> ".join(res) + " -> (cycles back...)"

cll = CircularLinkedList()
for x in [100, 200, 300]: cll.append(x)
print("Circular Display (showing 7 steps):", cll.display())
```

## 5. Algorithmic Pattern 1: Reversing a Linked List

Reverses pointers in-place in $O(n)$ time and $O(1)$ space using 3 pointers (`prev`, `curr`, `next_node`).


```python
# Iterative Reversal
def reverse_linked_list(head):
    prev = None
    curr = head
    while curr:
        next_node = curr.next  # 1. Save next node
        curr.next = prev       # 2. Reverse pointer
        prev = curr            # 3. Move prev forward
        curr = next_node       # 4. Move curr forward
    return prev                # New head

# Testing Reversal
head = Node(1)
head.next = Node(2)
head.next.next = Node(3)
head.next.next.next = Node(4)

def print_nodes(h):
    res = []
    while h:
        res.append(str(h.val))
        h = h.next
    return " -> ".join(res)

print("Before reversal:", print_nodes(head))
new_head = reverse_linked_list(head)
print("After reversal :", print_nodes(new_head))
```

## 6. Algorithmic Pattern 2: Fast & Slow Pointers (Floyd's Cycle Detection)

Uses two pointers moving at different speeds (`slow` moves 1 step, `fast` moves 2 steps).
If a cycle exists, `fast` and `slow` will meet inside the cycle in $O(n)$ time and $O(1)$ space.


```python
def has_cycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True  # Cycle detected!
    return False

# Create list with cycle: 1 -> 2 -> 3 -> 4 -> (points back to 2)
n1, n2, n3, n4 = Node(1), Node(2), Node(3), Node(4)
n1.next = n2; n2.next = n3; n3.next = n4; n4.next = n2

print("Has cycle?", has_cycle(n1))
n4.next = None  # Remove cycle
print("Has cycle after removal?", has_cycle(n1))
```

## 7. Algorithmic Pattern 3: Finding the Middle Node

When `fast` reaches the end of the list, `slow` is exactly at the middle node.


```python
def find_middle_node(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow.val if slow else None

# List 1 -> 2 -> 3 -> 4 -> 5
n1 = Node(1); n1.next = Node(2); n1.next.next = Node(3)
n1.next.next.next = Node(4); n1.next.next.next.next = Node(5)

print("Middle node of [1, 2, 3, 4, 5]:", find_middle_node(n1))
```

## 8. Algorithmic Pattern 4: Merging Two Sorted Linked Lists

Merges two sorted lists in-place using a dummy head node in $O(n + m)$ time and $O(1)$ space.


```python
def merge_two_lists(l1, l2):
    dummy = Node(0)
    tail = dummy
    
    while l1 and l2:
        if l1.val <= l2.val:
            tail.next = l1
            l1 = l1.next
        else:
            tail.next = l2
            l2 = l2.next
        tail = tail.next
        
    tail.next = l1 if l1 else l2
    return dummy.next

# L1: 1 -> 3 -> 5
l1 = Node(1); l1.next = Node(3); l1.next.next = Node(5)
# L2: 2 -> 4 -> 6
l2 = Node(2); l2.next = Node(4); l2.next.next = Node(6)

merged_head = merge_two_lists(l1, l2)
print("Merged Sorted Lists:", print_nodes(merged_head))
```

## 9. Algorithmic Pattern 5: Intersection of Two Linked Lists

Finds the intersection node of two lists using two pointers.  
When pointer `p1` reaches the end of list A, redirect it to head B.  
When `p2` reaches end of list B, redirect it to head A.  
They will align and meet at the intersection node after at most $a + b$ steps!


```python
def get_intersection_node(headA, headB):
    if not headA or not headB: return None
    p1, p2 = headA, headB
    while p1 != p2:
        p1 = p1.next if p1 else headB
        p2 = p2.next if p2 else headA
    return p1  # Either intersection node or None

# Shared tail: 8 -> 10
shared = Node(8); shared.next = Node(10)
# List A: 1 -> 2 -> shared
headA = Node(1); headA.next = Node(2); headA.next.next = shared
# List B: 5 -> shared
headB = Node(5); headB.next = shared

intersect = get_intersection_node(headA, headB)
print("Intersection Node Value:", intersect.val if intersect else None)
```

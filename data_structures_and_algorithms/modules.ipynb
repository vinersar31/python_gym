## 1. What is a Module?

In Python, a **module** is simply a file ending with `.py` that contains Python code.
Modules allow you to:
- **Organize code**: Group related functions, classes, and variables together.
- **Reuse code**: Import and use functions across multiple projects without duplicating code.
- **Manage namespaces**: Avoid name collisions by prefixing functions with the module name.

Python comes with a rich set of built-in modules called the **Standard Library**
(e.g., `math`, `sys`, `os`, `json`, `random`, `datetime`).


````python
# Importing a standard library module
import math

# Accessing functions and constants defined inside the 'math' module
print(f"math.pi        = {math.pi}")
print(f"math.e         = {math.e}")
print(f"math.sqrt(16)  = {math.sqrt(16)}")
print(f"math.factorial = {math.factorial(5)}")
print(f"math.gcd(24,36)= {math.gcd(24, 36)}")
````

## 2. Creating & Importing a Custom Module

Let's create a custom Python file called `calculator.py` using Jupyter's `%%writefile` magic command.
This file will act as our custom module.


````python
# Write a custom module 'calculator.py' to disk
%%writefile calculator.py
"""A simple calculator module for arithmetic operations."""

AUTHOR = "Python Developer"
VERSION = "1.0.0"

def add(a, b):
    """Return the sum of a and b."""
    return a + b

def subtract(a, b):
    """Return the difference of a and b."""
    return a - b

def multiply(a, b):
    """Return the product of a and b."""
    return a * b

def divide(a, b):
    """Return the quotient of a and b."""
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero")
    return a / b
````

````python
# Import and use the custom module
import calculator

print(f"Module Name   : {calculator.__name__}")
print(f"Module Author : {calculator.AUTHOR}")
print(f"Module Version: {calculator.VERSION}")
print()
print(f"add(10, 5)     = {calculator.add(10, 5)}")
print(f"subtract(10, 5)= {calculator.subtract(10, 5)}")
print(f"multiply(10, 5)= {calculator.multiply(10, 5)}")
print(f"divide(10, 5)  = {calculator.divide(10, 5)}")
````

## 3. Import Statement Variants

Python offers several ways to import code from a module:

| Import Syntax | Description | Example |
|---------------|-------------|---------|
| `import module` | Imports module, access via `module.name` | `import math; math.sqrt(4)` |
| `import module as alias` | Imports module under a custom alias | `import numpy as np` |
| `from module import name` | Imports specific attributes directly into local namespace | `from math import pi, sqrt` |
| `from module import name as alias` | Imports specific attributes with local aliases | `from math import sqrt as square_root` |
| `from module import *` | Imports **all** public attributes (discouraged) | `from math import *` |


````python
# 1. import with alias (renaming module)
import calculator as calc

print(f"calc.add(20, 30) = {calc.add(20, 30)}")
````

````python
# 2. from module import specific items
from calculator import add, multiply

print(f"add(15, 25)      = {add(15, 25)}")
print(f"multiply(4, 6)   = {multiply(4, 6)}")
````

````python
# 3. from module import specific item with alias
from calculator import divide as div

print(f"div(100, 4)      = {div(100, 4)}")
````

````python
# 4. from module import * (Wildcard Import — Use with Caution!)
# Wildcard imports pollute the local namespace and can overwrite existing variables.
from calculator import *

print(f"AUTHOR           = {AUTHOR}")
print(f"subtract(50, 20) = {subtract(50, 20)}")
````

## 4. Module Executable Code & `__name__ == '__main__'`

When a Python file is executed directly, Python sets the special variable `__name__` to `'__main__'`.  
When the file is **imported** as a module into another script, `__name__` is set to the file's module name.

This allows a module to serve a dual purpose:
1. As a **reusable library** when imported.
2. As a **standalone script/test runner** when executed directly.

```python
if __name__ == '__main__':
    # Code here runs ONLY when executed directly as a script
    # Runs tests, CLI commands, or demo logic
```


````python
# Create a module with a main block
%%writefile greeter.py
def greet_user(name):
    return f"Hello, {name}! Welcome to Python modules."

def farewell_user(name):
    return f"Goodbye, {name}! Have a great day."

# Test / CLI block
if __name__ == '__main__':
    print("[RUNNING STANDALONE SCRIPT]")
    print(greet_user("Developer"))
    print(farewell_user("Developer"))
````

````python
# Run greeter.py as a script via terminal command
import subprocess

output = subprocess.check_output(["python", "greeter.py"], text=True)
print("Output when executed directly as script:")
print(output)
````

````python
# Import greeter.py as a module in Python
# Notice that the [RUNNING STANDALONE SCRIPT] block does NOT run upon import!
import greeter

print(f"Imported greeter.__name__ = {greeter.__name__!r}")
print(greeter.greet_user("Alice"))
````

## 5. The Module Search Path (`sys.path`)

When you statement `import my_module`, Python searches for the file in the following order:
1. The **current directory** (or script location).
2. Directories listed in the `PYTHONPATH` environment variable.
3. **Standard library** installation directories.
4. Third-party packages inside `site-packages`.

This search path is stored in `sys.path` as a list of directory paths.


````python
import sys

print("Python Module Search Path (sys.path):")
for i, path in enumerate(sys.path[:6]):
    print(f"  [{i}] {path}")
````

````python
# Dynamically adding a custom directory to sys.path
import sys
import os

custom_dir = os.path.abspath("./my_modules")
os.makedirs(custom_dir, exist_ok=True)

# Add custom_dir to search path if not present
if custom_dir not in sys.path:
    sys.path.append(custom_dir)

print(f"Custom directory added: {custom_dir}")
print(f"Is in sys.path? {custom_dir in sys.path}")
````

## 6. Module Caching & Reloading (`importlib.reload`)

For efficiency, Python **caches imported modules** in `sys.modules` dictionary.
Subsequent `import` statements for the same module do **not** re-read or re-execute the module file!

If you edit a module `.py` file while working in Jupyter or an interactive session,
you must use `importlib.reload()` to load the updated code.


````python
# Create initial version of configurable module
%%writefile config_module.py
CONFIG_VAL = 100
def get_status():
    return f"Status V1: config = {CONFIG_VAL}"
````

````python
# First import
import config_module
print(config_module.get_status())
````

````python
# Modify config_module.py on disk
%%writefile config_module.py
CONFIG_VAL = 200
def get_status():
    return f"Status V2 (UPDATED!): config = {CONFIG_VAL}"
````

````python
# Plain import does NOT update because module is cached in sys.modules
import config_module
print("Before reload :", config_module.get_status())

# Force reload using importlib.reload()
import importlib
importlib.reload(config_module)
print("After reload  :", config_module.get_status())
````

## 7. Module Introspection (`dir()`, `__file__`, `__doc__`)

Python provides built-in attributes and functions to inspect imported modules:

| Introspection Tool | Description |
|--------------------|-------------|
| `dir(module)` | Returns a list of all defined attributes and functions |
| `module.__name__` | Returns the module's name string |
| `module.__file__` | Returns the file path of the module |
| `module.__doc__` | Returns the module docstring |
| `help(module)` | Displays interactive help and documentation |


````python
import calculator

print(f"module.__name__ : {calculator.__name__}")
print(f"module.__doc__  : {calculator.__doc__}")
print(f"module.__file__ : {calculator.__file__}")

print("\ndir(calculator) public attributes:")
public_attrs = [attr for attr in dir(calculator) if not attr.startswith("__")]
print(public_attrs)
````

## 8. Standard Library Modules Showcase

Python's slogan is **"Batteries Included"**. Here are several indispensable built-in modules:


````python
# 1. random module — random numbers and choices
import random
random.seed(42)

print(f"random float [0,1) : {random.random():.4f}")
print(f"random int [1, 6]  : {random.randint(1, 6)}")
fruits = ["apple", "banana", "cherry", "date"]
print(f"random choice      : {random.choice(fruits)}")
print(f"random sample (2)  : {random.sample(fruits, 2)}")
````

````python
# 2. datetime module — date and time handling
from datetime import datetime, timedelta

now = datetime.now()
tomorrow = now + timedelta(days=1)
formatted = now.strftime("%Y-%m-%d %H:%M:%S")

print(f"Current time : {formatted}")
print(f"Tomorrow     : {tomorrow.strftime('%Y-%m-%d')}")
````

````python
# 3. json module — JSON serialization & parsing
import json

user_data = {"id": 101, "name": "Alice", "skills": ["Python", "SQL"], "active": True}
json_str = json.dumps(user_data, indent=2)
print("JSON String:\n", json_str)

parsed = json.loads(json_str)
print(f"Parsed object type: {type(parsed)}, active = {parsed['active']}")
````

````python
# 4. collections module — specialized container datatypes
from collections import Counter, defaultdict, namedtuple

# Counter
counts = Counter("abracadabra")
print(f"Counter top 2: {counts.most_common(2)}")

# defaultdict
dd = defaultdict(list)
dd["evens"].append(2)
dd["evens"].append(4)
print(f"defaultdict  : {dict(dd)}")

# namedtuple
Point = namedtuple("Point", ["x", "y"])
pt = Point(10, 20)
print(f"namedtuple   : x={pt.x}, y={pt.y}")
````

## 9. Python Packages & `__init__.py`

A **package** is a directory containing Python modules and a special file named `__init__.py`.

Structure of a package:
```
mypackage/
│── __init__.py      # Marks directory as a Python package (can be empty)
│── formatter.py    # Submodule 1
└── validator.py    # Submodule 2
```

Packages allow hierarchical structuring of module namespaces using dot notation:
`import mypackage.formatter`


````python
# Create package directory structure
import os

os.makedirs("mypackage", exist_ok=True)
````

````python
# Create mypackage/__init__.py
%%writefile mypackage/__init__.py
"""mypackage Initialization file."""
PACKAGE_NAME = "MyPackage"
PACKAGE_VERSION = "1.0.0"

print(f"[{PACKAGE_NAME} initialized]")
````

````python
# Create mypackage/formatter.py
%%writefile mypackage/formatter.py
def uppercase_text(text):
    return text.upper()

def title_text(text):
    return text.title()
````

````python
# Create mypackage/validator.py
%%writefile mypackage/validator.py
def is_valid_email(email):
    return "@" in email and "." in email

def is_positive_number(n):
    return isinstance(n, (int, float)) and n > 0
````

````python
# Import and use submodules from the package
import mypackage.formatter as fmt
import mypackage.validator as val

print(fmt.uppercase_text("python modules and packages"))
print(f"email valid? : {val.is_valid_email('alice@example.com')}")
print(f"number > 0?  : {val.is_positive_number(42)}")
````

## 10. Package Export Control (`__all__`)

The `__all__` variable inside `__init__.py` or a module defines what is exported when a user types:
`from mypackage import *`

This allows package authors to control their public API surface and hide internal helper functions.


````python
# Update mypackage/__init__.py with __all__ and re-exports
%%writefile mypackage/__init__.py
from .formatter import uppercase_text, title_text
from .validator import is_valid_email

__all__ = ["uppercase_text", "title_text", "is_valid_email"]
````

````python
# Test Package import with __all__
import importlib
import mypackage
importlib.reload(mypackage)

from mypackage import *

print(uppercase_text("hello from package export"))
print(f"valid email? {is_valid_email('user@domain.com')}")
````

## 11. Quick Reference & Best Practices

### Best Practices
1. **PEP 8 Import Order**: Group imports in this order:
   - Standard library imports (e.g., `import os`, `import sys`)
   - Related third-party imports (e.g., `import numpy`, `import pandas`)
   - Local application/library specific imports
2. **Avoid Wildcard Imports**: Avoid `from module import *` — it obscures origin of names.
3. **Use `if __name__ == '__main__':`** for test/demo code in module files.
4. **Module docstrings**: Always include a docstring at the top of a module.
5. **Circular Imports**: Avoid circular dependencies where Module A imports B and B imports A.


````python
# ==================================================================
# PYTHON MODULES & PACKAGES – QUICK REFERENCE
# ==================================================================

# --- Basic Imports ---
import math                       # math.sqrt()
import math as m                  # m.sqrt()
from math import sqrt, pi         # sqrt(), pi
from math import sqrt as square_root # square_root()

# --- Inspect Module ---
import calculator
print(f"Name : {calculator.__name__}")
print(f"File : {calculator.__file__}")
print(f"Attrs: {[a for a in dir(calculator) if not a.startswith('__')]}")

# --- Reloading ---
import importlib
importlib.reload(calculator)

# --- Search Path ---
import sys
print(f"Search paths count: {len(sys.path)}")
````

## 1. Stack Concepts & Operations

| Operation | Method | Time Complexity | Space Complexity |
|-----------|--------|-----------------|------------------|
| **Push** | Add item to top | $O(1)$ | $O(1)$ |
| **Pop** | Remove top item | $O(1)$ | $O(1)$ |
| **Peek / Top** | Inspect top item | $O(1)$ | $O(1)$ |
| **Is Empty** | Check if empty | $O(1)$ | $O(1)$ |


```python
# Basic Stack operations using collections.deque
from collections import deque

stack = deque()
stack.append("Page 1")
stack.append("Page 2")
stack.append("Page 3")

print("Stack state :", list(stack))
print("Peek top    :", stack[-1])
print("Popped item :", stack.pop())
print("After pop   :", list(stack))
```

## 2. Stack Implementations (`list` vs. `collections.deque`)

- **Python `list`**: Works as a stack via `.append()` and `.pop()`, but suffers from dynamic array reallocation spikes.
- **`collections.deque`**: Double-ended queue implemented as a doubly linked block array, guaranteeing consistent $O(1)$ operations.

## 3. Building a Custom Stack Class (Linked List Backend)


```python
class Node:
    def __init__(self, val):
        self.val = val
        self.next = None

class Stack:
    def __init__(self):
        self.top_node = None
        self._size = 0

    def push(self, val):
        new_node = Node(val)
        new_node.next = self.top_node
        self.top_node = new_node
        self._size += 1

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        val = self.top_node.val
        self.top_node = self.top_node.next
        self._size -= 1
        return val

    def peek(self):
        if self.is_empty():
            raise IndexError("peek from empty stack")
        return self.top_node.val

    def is_empty(self):
        return self.top_node is None

    def __len__(self):
        return self._size

# Testing Custom Stack
st = Stack()
st.push(10); st.push(20); st.push(30)
print("Peek:", st.peek())
print("Pop :", st.pop())
print("Len :", len(st))
```

## 5. Algorithmic Pattern 1: Balanced Parentheses Validation


```python
def is_valid_parentheses(s):
    matching = {')': '(', ']': '[', '}': '{'}
    stack = []
    for char in s:
        if char in "({[":
            stack.append(char)
        elif char in ")}]":
            if not stack or stack[-1] != matching[char]:
                return False
            stack.pop()
    return len(stack) == 0

print("({[]}) valid?", is_valid_parentheses("({[]})"))
print("([)]   valid?", is_valid_parentheses("([)]"))
```

## 6. Algorithmic Pattern 2: Monotonic Stack (Next Greater Element)

Finds the next greater element for each item in an array in $O(n)$ time using a monotonic stack.


```python
def next_greater_element(nums):
    res = [-1] * len(nums)
    stack = []  # Stores indices
    for i, num in enumerate(nums):
        while stack and nums[stack[-1]] < num:
            idx = stack.pop()
            res[idx] = num
        stack.append(i)
    return res

nums = [2, 1, 2, 4, 3]
print("Nums        :", nums)
print("Next Greater:", next_greater_element(nums))
```

## 7. Algorithmic Pattern 3: Evaluate Reverse Polish Notation (RPN)


```python
def eval_rpn(tokens):
    stack = []
    for token in tokens:
        if token in "+-*/":
            b, a = stack.pop(), stack.pop()
            if token == '+': stack.append(a + b)
            elif token == '-': stack.append(a - b)
            elif token == '*': stack.append(a * b)
            elif token == '/': stack.append(int(a / b))  # Truncate toward zero
        else:
            stack.append(int(token))
    return stack[0]

rpn_expression = ["2", "1", "+", "3", "*"]  # (2 + 1) * 3 = 9
print("RPN Evaluation result:", eval_rpn(rpn_expression))
```

## 8. Algorithmic Pattern 4: Min Stack ($O(1)$ Minimum Element Retrieval)


```python
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, val: int) -> None:
        self.stack.append(val)
        current_min = min(val, self.min_stack[-1] if self.min_stack else val)
        self.min_stack.append(current_min)

    def pop(self) -> None:
        self.stack.pop()
        self.min_stack.pop()

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        return self.min_stack[-1]

ms = MinStack()
ms.push(-2); ms.push(0); ms.push(-3)
print("Min val:", ms.getMin())  # Returns -3
ms.pop()
print("Top val:", ms.top())     # Returns 0
print("Min val:", ms.getMin())  # Returns -2
```
